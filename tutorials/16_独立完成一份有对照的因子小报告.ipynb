## 4. 从数据到表

### 4.1 读数据、算标签



```python

import pandas as pd

df = pd.read_csv("data/rb_events.csv", parse_dates=["day", "signal"])
df["move"] = df["entry"] - df["exit"]

print("事件数", len(df), "交易日", df["day"].nunique())
print(df[["score", "move"]].describe().round(3))
```


**读数**：平均顺向变化 −0.36 点；普通下破整体不是赚钱的做空信号。报告里要先把这句话写出来，再谈分组。

### 4.2 训练段定阈值，测试段只应用



```python

train = df[df["day"] <= pd.Timestamp("2026-08-31")]
test = df[df["day"] > pd.Timestamp("2026-08-31")]
lo, hi = train["score"].quantile([1 / 3, 2 / 3])

def group_table(d):
    d = d.copy()
    d["group"] = pd.cut(d["score"], [-float("inf"), lo, hi, float("inf")],
                        labels=["low", "mid", "high"])
    return d.groupby("group", observed=True).agg(
        n=("move", "size"), days=("day", "nunique"),
        mean=("move", "mean"), median=("move", "median"))

print("训练段阈值", round(lo, 3), round(hi, 3))
print("测试段")
print(group_table(test).round(3))
```


**读数**：测试段高分组 −0.152 点、低分组 −1.000 点；高分组少亏，但两组都是负的。

### 4.3 同日配对与区间



```python

import numpy as np

test = test.copy()
test["group"] = pd.cut(test["score"], [-float("inf"), lo, hi, float("inf")],
                       labels=["low", "mid", "high"])
pivot = test.pivot_table(index="day", columns="group", values="move",
                         aggfunc="mean", observed=True)
paired = (pivot["high"] - pivot["low"]).dropna()

rng = np.random.default_rng(16)
boot = rng.choice(paired.to_numpy(), size=(5000, len(paired)), replace=True).mean(axis=1)

print("配对交易日", len(paired), "高减低", round(paired.mean(), 3))
print("95% 区间", tuple(round(float(v), 3) for v in np.quantile(boot, [0.025, 0.975])))
```


**读数**：14 个配对日，高减低 +0.702 点，95% 区间 [−0.946, 2.637]，跨 0。

### 4.4 成本账



```python

for cost in [2, 4]:
    high_net = (test.loc[test["group"] == "high", "move"].mean() - cost) * 10 - 6
    low_net = (test.loc[test["group"] == "low", "move"].mean() - cost) * 10 - 6
    print("每边", cost, "点：高分组", round(high_net, 2), "低分组", round(low_net, 2))
```


**读数**：2 点成本下，高分组每笔 −27.52 元，低分组 −36.00 元；4 点成本下更差。

## 5. 参考报告（文字版）

### 5.1 问题与假设

**问题**：普通下破前“跌得顺”的合约，下破后 10 分钟是否更容易延续？

**假设**：若下跌是“一路卖下来”的，说明卖压持续；若只是“最后一根砸破”，更容易反抽。可反驳的预测：下行效率高的组，未来 10 分钟顺向跌幅应大于低效率组。

### 5.2 数据与方法

- 数据：RB2701 历史分钟线整理出的 256 次事件、72 个交易日；
- 事件：5 分钟收盘跌破前两根完整 5 分钟线最低价；
- 因子：突破前 5 个收盘价的下行效率；
- 标签：`entry - exit`（做空方向，正数=下跌）；
- 切分：2026-08-31 前为训练段，之后为测试段；
- 阈值：训练段 1/3、2/3 分位（−0.20 / 0.33），测试段原样应用；
- 成本：每点 10 元，每边 2 点价差 + 6 元手续费；
- 不确定性：同日高减低配对 + 5,000 次 bootstrap。

### 5.3 结果

| 组 | 事件 | 交易日 | 平均顺向（点） | 中位数 | 2 点成本净利（元） |
|---|---:|---:|---:|---:|---:|
| 低 | 33 | 23 | −1.000 | −1.0 | −36.00 |
| 中 | 38 | 23 | −0.421 | −0.5 | −24.21 |
| 高 | 33 | 18 | −0.152 | −1.0 | −27.52 |

- 同日高减低：+0.702 点；
- 95% bootstrap 区间：[−0.946, 2.637]；
- 配对交易日：14 天。

### 5.4 结论

高分组比低分组“少亏”约 0.85 点，但：

1. 两组平均都是负的；
2. 扣 2 点价差 + 6 元手续费后，两组都亏；
3. 高减低区间跨 0，14 个配对日不足以支持“稳定有效”。

**研究决定：不交易。** 保留为观察项；扩样本、加盘口、做前向观察后再评估。

### 5.5 限制

- 只有分钟价格，没有盘口；entry/exit 不是可成交回报；
- 样本已参与研究，不是全新样本；
- 14 个配对日，区间很宽；
- 没有做多重检验/品种扩展；
- 成本只按固定 2/4 点，未建模冲击成本。

## 7. 常见错误与检查清单

### 7.1 五个常见错误

1. **把“方向对”当“赚钱”**。高分组少亏，但扣费后仍亏。
2. **用全样本分位**。阈值偷看未来。
3. **只看均值**。中位数、样本量、区间都要看。
4. **不做同日配对**。日期效应会被当成因子效应。
5. **结论写“有效”**。没有样本外/成本/区间，不能写有效。

### 7.2 交卷前检查

- [ ] 问题能用一句话说清；
- [ ] 数据/字段/样本期/独立交易日；
- [ ] 标签方向、窗口、重叠处理；
- [ ] 训练/测试切分，阈值来源；
- [ ] 全做/筛过/被拒三行；
- [ ] 成本含乘数、双边、价差；
- [ ] 区间/样本量/限制；
- [ ] 结论能被反例推翻。

---

## 本章小结

一份“能复查”的报告，不是“结论漂亮”，而是：

- 别人能按你的规则复算；
- 对照公平（全做/筛过/被拒、同日配对）；
- 成本算到一手净利；
- 不确定性有区间/样本量；
- 结论承认“不可交易”也是一种结论。

**如果这份报告只能留一句话**：先证明“扣费后还剩下什么”，再谈“因子有没有效”。
