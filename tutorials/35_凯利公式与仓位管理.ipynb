## 1. 二元赌局：凯利公式怎么推出来

### 1.1 设定

每局押总资金的 $f$：

- 赢：资金乘以 $1+bf$（净赚 $bf$）；
- 输：资金乘以 $1-f$（亏掉 $f$）；
- 胜率 $p$，败率 $q=1-p$，赔率 $b:1$。

长期增长率（对数财富的期望）：

$$g(f)=p\ln(1+bf)+q\ln(1-f).$$

### 1.2 求导

$$\frac{dg}{df}=\frac{pb}{1+bf}-\frac{q}{1-f}=0.$$

交叉相乘：

$$pb(1-f)=q(1+bf)\quad\Rightarrow\quad f^*=\frac{pb-q}{b}=\frac{p(b+1)-1}{b}.$$

**公式拆解**
- **符号**：$f^*$ 是每局押上的资金比例，$p$ 是胜率、$q=1-p$ 是败率，$b$ 是赔率即赢时净赚本金的倍数，时间顺序是每局按当前资金的 $f$ 下注、局后按输赢更新资金。
- **结构**：分子 $pb-q$ 是每 1 元押注的期望净收益，分母 $b$ 把它换算成应押的本金比例，也可以读成 $p(b+1)-1$ 除以 $b$ 的净 edge，赔率越高同样胜率就能押得越多。
- **代入**：$p=0.6,b=1$ 得 $f^*=(0.6-0.4)/1=0.2$ 即押 20%，若 $p=0.45,b=2$ 则 $f^*=(0.9-0.55)/2=0.175$。
- **边界**：$pb\le q$ 时公式给出 0 或负值、应截断为 0，$f^*>1$ 时数学上要求杠杆而实盘未必允许。
- **用途**：给出长期增长最快的下注比例，是仓位管理的理论基准；它不是重仓许可，$p,b$ 估错就会过度下注。

若 $pb-q\le0$，最优是 $f^*=0$（不下注）。

### 1.3 手算

- $p=0.6,b=1$：$f^*=(0.6-0.4)/1=0.2$；
- $p=0.55,b=1$：$f^*=0.1$；
- $p=0.5,b=1$：$f^*=0$；
- $p=0.6,b=2$：$f^*=(1.2-0.4)/2=0.4$。

**赔率越高，同样胜率下可以押更多；但胜率低于盈亏平衡点，凯利是 0。**

### 1.4 代码：二元凯利表



```python

def kelly_binary(p, b):
    q = 1 - p
    return max((p * b - q) / b, 0.0)

for p, b in [(0.6, 1), (0.55, 1), (0.5, 1), (0.6, 2), (0.45, 2)]:
    print(f"p={p}, b={b}: Kelly={kelly_binary(p, b):.2%}")
```


**读数**：$p=0.6,b=2$ 时押 40%；$p=0.45,b=2$ 时 $pb-q=0.9-0.55=0.35>0$，押 17.5%。**正期望不等于“重仓”，凯利给的是一个使对数增长最大的比例。**

**常见错误**：

1. 把“期望收益/赔率”当凯利。
2. 忘了 $f^*$ 可能大于 1（需要杠杆）。
3. 负 edge 还硬算，得到负仓位。

**停下来想一想**：$p=0.6,b=1$，凯利 20%；为什么很多人实际只押 5%–10%？

<details><summary>对照你的思路</summary>

因为 $p$ 是估出来的，估错会把 20% 放大成灾难；全凯利的回撤很大；还有尾部、相关性、保证金。实务用分数凯利，把“参数误差/回撤”换成“增长略低但活得久”。

</details>

## 2. 连续收益：$f^*=\mu/\sigma^2$

### 2.1 推导直觉

若仓位 $f$ 的收益近似 $f\mu-\frac12 f^2\sigma^2$，对数增长最大化：

$$\max_f\ f\mu-\frac12 f^2\sigma^2\quad\Rightarrow\quad f^*=\frac{\mu}{\sigma^2}.$$

**公式拆解**
- **符号**：$\mu$ 是超额收益的期望（年化小数），$\sigma$ 是同频率的波动率，$f^*$ 是仓位倍数、大于 1 表示需要杠杆，$\mu,\sigma$ 都只能用历史或预测估计。
- **结构**：目标 $f\mu-\frac12f^2\sigma^2$ 是收益减半方差，第一项线性增、第二项按平方减，倒 U 形的顶点就是 $f^*=\mu/\sigma^2$，注意分母是方差不是波动率。
- **代入**：$\mu=8\%$、$\sigma=20\%$ 得 $f^*=0.08/0.04=2.0$ 即要求 2 倍杠杆，若 $\sigma$ 降到 10% 则 $f^*=8$，半凯利取 0.5 才回到满仓。
- **边界**：$\mu\le0$ 时不应配仓，$\sigma\to0$ 时 $f^*$ 发散，公式还假设收益正态、参数已知且可无限杠杆。
- **用途**：它是风险厌恶 $\lambda=1$ 时的均值方差组合、也是分数凯利的基准；不能直接执行，实务要先乘折扣系数并检查杠杆上限。

这是“连续凯利”：预期超额收益除以方差。

### 2.2 例子

$\mu=8\%$、$\sigma=20\%$：

$$f^*=\frac{0.08}{0.04}=2.0.$$

**凯利要求 2 倍杠杆。** 这就是为什么“全凯利”在实盘几乎不可行：它假设 $\mu,\sigma$ 已知、收益正态、可无限杠杆。

### 2.3 代码：连续凯利



```python

mu, sigma = 0.08, 0.20
f_star = round(mu / sigma ** 2, 2)
print("连续凯利", f_star)
print("半凯利", f_star / 2, "四分之一凯利", f_star / 4)
```


**读数**：2 倍、1 倍、0.5 倍。若杠杆上限是 1 倍，全凯利根本不可执行；半凯利也可能超限。

**常见错误**：

1. 用收益率而不是超额收益。
2. 用年化 $\sigma$ 却忘了 $\mu$ 也是年化。
3. 忽略杠杆/保证金约束。

**停下来想一想**：$\mu=8\%,\sigma=20\%$，凯利 2 倍；若 $\mu$ 的估计误差是 $\pm4\%$，仓位区间是多少？

<details><summary>对照你的思路</summary>

$f=\mu/\sigma^2$，$\mu$ 从 4% 到 12% 对应 1 倍到 3 倍；参数误差直接把仓位放大 3 倍。这就是分数凯利的理由。

</details>

## 3. 分数凯利：增长与回撤的交换

### 3.1 定义

$$f=c f^*,\qquad 0<c\le1.$$

- $c=1$：全凯利，增长最快，回撤最大；
- $c=1/2$：约 3/4 的增长，约 1/2 的波动；
- $c=1/4$：更稳，增长更慢。

### 3.2 模拟

用 $p=0.6,b=1$，200 局，比较 $f=0,0.1,0.2,0.4$ 的终端财富、最大回撤、破产概率。

### 3.3 代码：路径模拟



```python

import numpy as np

def simulate(p, b, f, n_bets=200, n_paths=5000, seed=35):
    rng = np.random.default_rng(seed)
    wins = rng.random((n_paths, n_bets)) < p
    returns = np.where(wins, 1 + b * f, 1 - f)
    wealth = np.cumprod(returns, axis=1)
    terminal = wealth[:, -1]
    peak = np.maximum.accumulate(wealth, axis=1)
    max_dd = (wealth / peak - 1).min(axis=1)
    ruin = (wealth.min(axis=1) <= 0.1).mean()
    return terminal, max_dd, ruin

for f in [0.0, 0.1, 0.2, 0.4]:
    terminal, max_dd, ruin = simulate(0.6, 1, f)
    print(f"f={f:.1f}: 中位终值={np.median(terminal):7.2f}, "
          f"5%分位={np.quantile(terminal,0.05):6.2f}, "
          f"中位最大回撤={np.median(max_dd):6.2%}, 破产率={ruin:6.2%}")
```


**读数**：$f=0.2$ 的中位终值最高；$f=0.4$（两倍凯利）中位终值下降、回撤和破产率上升。**超过凯利，增长反而下降。**

**常见错误**：

1. 只看均值，不看回撤/破产。
2. 用“历史最大回撤”代替分布。
3. 把分数凯利当“保守”，不检查杠杆上限。

**停下来想一想**：为什么“超过全凯利”会降低长期增长？

<details><summary>对照你的思路</summary>

对数增长 $g(f)$ 在 $f^*$ 处最大，超过后二阶项为负；波动/回撤的复利拖累超过收益。数学上 $g''(f^*)<0$，所以“押更大”不是“赚更多”。

</details>

## 4. 参数误差：估错一点，仓位偏很多

### 4.1 胜率的抽样误差

若真实 $p=0.55$，用 100 笔估计：

$$\hat p\sim N\left(0.55,\frac{0.55\times0.45}{100}\right),$$

标准差约 5 个百分点。对应 $f=2\hat p-1$ 的标准差约 10 个百分点；有时估出负 edge，凯利为 0。

### 4.2 代码：$f$ 的分布



```python

rng = np.random.default_rng(3501)
true_p, n = 0.55, 100
p_hat = rng.binomial(n, true_p, 20000) / n
f_hat = np.maximum(2 * p_hat - 1, 0)
print("p_hat 均值/标准差", round(p_hat.mean(), 3), round(p_hat.std(ddof=1), 3))
print("f 均值/标准差", round(f_hat.mean(), 3), round(f_hat.std(ddof=1), 3))
print("f=0 的比例", round((f_hat == 0).mean(), 3))
print("f 的 5%/95% 分位", tuple(round(float(v), 3) for v in np.quantile(f_hat, [0.05, 0.95])))
```


**读数**：真实凯利 0.10，但估计出的 $f$ 均值约 0.10、标准差约 0.10；5% 分位接近 0，95% 分位约 0.27。**用点估计直接上全凯利，会把 10% 的仓位做到 27%。**

**常见错误**：

1. 用点估计当真实参数。
2. 不报告参数不确定性。
3. 样本越短，越敢重仓。

**停下来想一想**：为什么“分数凯利”能部分解决参数误差？

<details><summary>对照你的思路</summary>

分数凯利把 $f$ 缩小，等于对 $\mu$ 的估计误差做保守处理；它牺牲一部分增长，换取对估错/尾部/相关性的缓冲。它不是数学最优，是“稳健最优”。

</details>

## 5. 多资产凯利：和均值-方差的关系

### 5.1 公式

多资产连续凯利：

$$f^*=\Sigma^{-1}\mu.$$

**公式拆解**
- **符号**：$f^*$ 是各资产的仓位向量（可正可负、可大于 1），$\Sigma$ 是收益协方差矩阵，$\mu$ 是超额收益向量，$\Sigma^{-1}$ 要求矩阵可逆。
- **结构**：它是单资产 $f=\mu/\sigma^2$ 的矩阵版，$\Sigma^{-1}\mu$ 同时除以方差并扣除相关性，与均值方差 $w=\frac1\lambda\Sigma^{-1}\mu$ 只差一个常数。
- **代入**：$\mu=(8\%,4\%)$、$\sigma=(20\%,10\%)$、$\rho=0.3$ 时算出 $\Sigma^{-1}\mu=(1.54,3.08)$，即资产 1 配 154%、资产 2 配 308%，总杠杆约 4.6 倍。
- **边界**：$\Sigma$ 不可逆（共线或样本太短）时无解或解不唯一，负相关会给出负权重，$\mu$ 的小误差会被 $\Sigma^{-1}$ 放大。
- **用途**：多资产仓位的理论基准，解释相关资产为何不能各配一份；它不是可执行方案，实务要加约束、收缩或改用风险平价，分数凯利则相当于把风险厌恶调到 1 以上。

均值-方差（风险厌恶 $\lambda$）：

$$w=\frac{1}{\lambda}\Sigma^{-1}\mu.$$

**凯利就是 $\lambda=1$ 的均值-方差组合。** 分数凯利相当于把风险厌恶调高。

### 5.2 代码：两资产



```python

mu = np.array([0.06, 0.10])
cov = np.array([[0.04, 0.01], [0.01, 0.09]])
f = np.linalg.solve(cov, mu)
print("多资产凯利", f.round(3), "总杠杆", round(f.sum(), 3))
print("半凯利", (0.5 * f).round(3), "总杠杆", round(0.5 * f.sum(), 3))
```


**读数**：两资产凯利约 1.26/0.97，总杠杆 2.23 倍；半凯利约 0.63/0.49。**多资产凯利常给出高杠杆，因为协方差对冲会放大仓位。**

**常见错误**：

1. 不检查协方差矩阵条件数。
2. 把负权重当“不能做”，不检查卖空约束。
3. 不做杠杆/保证金上限。

**停下来想一想**：为什么多资产凯利的总杠杆可能比单资产更高？

<details><summary>对照你的思路</summary>

因为资产间相关性低/负，组合方差被分散，$\Sigma^{-1}\mu$ 会放大每个资产的仓位；总杠杆是绝对值之和，可能远大于 1。相关性估计错误时，这种放大很危险。

</details>

## 6. 实务约束：凯利只是起点

### 6.1 六条约束

1. **杠杆上限**：交易所/券商/内部限额；
2. **保证金**：期货/期权卖方占用；
3. **尾部**：跳空/流动性/相关性突变；
4. **参数误差**：用置信下界/分数凯利；
5. **回撤限额**：超过就减仓；
6. **容量**：规模越大，冲击成本越高。

## 本章小结

### 一张表

| 概念 | 公式 | 用途 |
|---|---|---|
| 二元凯利 | $f^*=(pb-q)/b$ | 赌局/离散信号 |
| 连续凯利 | $f^*=\mu/\sigma^2$ | 收益/波动 |
| 多资产凯利 | $f^*=\Sigma^{-1}\mu$ | 组合 |
| 分数凯利 | $c f^*$ | 稳健 |

### 八句话

1. 正期望不等于重仓。
2. 凯利最大化对数增长，不是最大化期望。
3. 全凯利回撤大，实务用分数。
4. 参数估错，仓位会成倍偏。
5. 连续凯利常要求高杠杆。
6. 多资产凯利会因分散放大总杠杆。
7. 凯利是 $\lambda=1$ 的均值-方差。
8. 杠杆/保证金/尾部/回撤是硬约束。
