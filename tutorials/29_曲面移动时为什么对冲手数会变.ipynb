## 1. 固定行权价 vs 固定相对行权价

### 1.1 两种假设

- **Sticky strike**：标的动，某个固定执行价的隐波不变；
- **Sticky delta/moneyness**：标的动，某个相对执行价 $k=\ln(K/F)$ 的隐波不变。

真实市场介于两者之间，还随期限/品种变化。

### 1.2 对 Delta 的影响

若隐波随标的动，期权价格对标的的“全导数”不再只是 $\partial V/\partial F$，还要加 $\partial V/\partial\sigma\cdot\partial\sigma/\partial F$。

### 1.3 代码：两种 Delta



```python

import numpy as np
from scipy.special import ndtr

def black76(F, K, T, vol, r=0.0):
    w = vol * np.sqrt(T)
    d1 = (np.log(F / K) + 0.5 * w * w) / w
    return np.exp(-r * T) * (F * ndtr(d1) - K * ndtr(d1 - w))

def smile(k):
    return 0.22 - 0.16 * k + 0.30 * k ** 2

F, K, T, h = 100.0, 105.0, 0.25, 1.0
fixed = (black76(F + h, K, T, smile(np.log(K / F))) - black76(F - h, K, T, smile(np.log(K / F)))) / (2 * h)
moved = (black76(F + h, K, T, smile(np.log(K / (F + h)))) - black76(F - h, K, T, smile(np.log(K / (F - h))))) / (2 * h)

print("固定行权价 Delta", round(fixed, 4))
print("随曲面移动 Delta", round(moved, 4))
print("差异", round(moved - fixed, 4))
```


**读数与应用**：固定行权价 Delta 与随曲面移动 Delta 不同；差异来自 Vega × 隐波对标的的斜率。对冲时要说清用的哪种假设。

**常见错误**：

1. 用固定行权价 Delta 对冲 sticky delta 的市场。
2. 忽略 Vega，只看 Delta。
3. 不同执行价用不同假设，组合不一致。

**停下来想一想**：为什么“标的涨、隐波也涨”会让看涨期权的实际 Delta 更大？

<details><summary>对照你的思路</summary>

看涨期权 Vega 为正；标的涨时，若曲面移动让隐波也涨，Vega 贡献额外收益，期权对标的的整体敏感度（全导数）大于固定隐波的 Delta。反之，隐波跌会削弱 Delta。

</details>

## 2. 全导数：Delta + Vega × 曲面斜率

### 2.1 公式

$$\frac{dV}{dF}=\frac{\partial V}{\partial F}+\frac{\partial V}{\partial\sigma}\frac{\partial\sigma}{\partial F}.$$

- 第一项：固定隐波的 Delta；
- 第二项：Vega × 隐波对标的的斜率。

**公式拆解**
- **符号**：$F$ 标的价，$V(F,\sigma)$ 期权价，$V_F$ 固定隐波 Delta，$V_\sigma$ Vega，$\sigma_F$ 曲面斜率。
- **结构**：$dV/dF=V_F+V_\sigma\sigma_F$：直接价格效应 + 曲面移动的额外 Delta。
- **代入**：Vega 18.38×0.001307 → 修正 0.024；Delta 0.3429→0.3668。
- **边界**：$\sigma_F=0$ 退化为固定 Delta；深度实值全导数可 $>1$。
- **用途**：sticky delta 对冲；不是模型 Delta，也不是方向预测。

### 2.2 曲面斜率

在 $k=\ln(K/F)$ 坐标下，固定 $k$ 意味着：

$$\frac{\partial\sigma}{\partial F}=-\frac{1}{F}\frac{\partial\sigma}{\partial k}.$$

**公式拆解**
- **符号**：$k=\ln(K/F)$ 相对行权价，$\sigma(k)$ 微笑，$\sigma_k$ 斜率。
- **结构**：固定 $k$ 时 $K=Fe^k$，链式法则 $\sigma_F=-\sigma_k/F$。
- **代入**：$k\approx0.0488$、$\sigma_k\approx-0.1307$ → $\sigma_F\approx0.001307$。
- **边界**：$F\to0^+$ 时发散；$\partial\sigma/\partial K$ 与 $\sigma_F$ 差 $K/F$。
- **用途**：由微笑斜率算全导数；不是 $d\sigma/dK$ 或常数。

所以 sticky delta 的全导数可以用微笑斜率算。

### 2.3 代码：Vega 与斜率



```python

b = 1e-4
vega = (black76(F, K, T, smile(np.log(K / F)) + b) - black76(F, K, T, smile(np.log(K / F)) - b)) / (2 * b)
slope = (smile(np.log(K / (F + h))) - smile(np.log(K / (F - h)))) / (2 * h)

print("Vega", round(vega, 4))
print("隐波对标的斜率", round(slope, 6))
print("全导数修正项", round(vega * slope, 6))
```


**读数与应用**：修正项 = Vega × 斜率；若斜率为负（标的涨、隐波跌），全导数小于固定隐波 Delta。

**常见错误**：

1. 把“隐波对执行价的斜率”直接当“隐波对标的的斜率”。
2. 不乘 Vega。
3. 不同期限用同一个斜率。

**停下来想一想**：为什么商品期权的偏斜通常比股指更陡？

<details><summary>对照你的思路</summary>

商品有供需/库存/交割/天气等非线性因素，下跌时波动率上升更明显；股指的偏斜也陡，但来源不同（杠杆/避险需求）。不同品种的曲面动态不能用同一假设。

</details>

## 3. 隐含密度：曲率在说什么

### 3.1 Breeden–Litzenberger

看涨价对执行价的二阶导，给出风险中性终值密度：

$$f(F_T)=e^{rT}\frac{\partial^2C}{\partial K^2}.$$

离散版：

$$f(K)\approx e^{rT}\frac{C(K+\Delta K)-2C(K)+C(K-\Delta K)}{(\Delta K)^2}.$$

**公式拆解**
- **符号**：$C(K)$ 同行权价看涨价，$r$ 无风险利率、$T$ 期限、$\Delta K$ 行权价间距。
- **结构**：$f(K)\approx e^{rT}[C(K+\Delta K)-2C(K)+C(K-\Delta K)]/(\Delta K)^2$，曲率即密度。
- **代入**：$K=100$ 处 $f\approx0.03669$（1/元），原始二阶差分约 $0.917$。
- **边界**：曲率为负（密度为负）先查报价；网格不均匀不能用中心差分。
- **用途**：反推隐含分布、查无套利；是风险中性密度，不是真实概率。

### 3.2 直觉

- 看涨价是“终值超过 K 的贴现概率”；
- 对 K 求一阶导，是“密度”的负值；
- 二阶导，是密度的正比。

曲率为负/密度为负，说明报价不满足凸性，可能有套利/数据问题。

### 3.3 代码：隐含密度



```python

import numpy as np

K = np.arange(80.0, 121.0, 5.0)
prices = np.array([black76(F, ki, T, smile(np.log(ki / F))) for ki in K])
d2 = np.diff(prices, 2) / 5.0 ** 2
density = np.exp(0.0 * T) * d2

print("二阶差分", d2.round(5))
print("隐含密度（未归一）", density.round(5))
```


**读数与应用**：二阶差分为正，密度非负；若某点为负，先查报价/时间戳/合约映射。离散网格越粗，密度越平滑但误差越大。

**常见错误**：

1. 用一阶差分当密度。
2. 不除以 $\Delta K^2$。
3. 不同到期/不同标的混在一起。

**停下来想一想**：为什么虚值看跌的报价会影响虚值看涨的密度？

<details><summary>对照你的思路</summary>

看涨/看跌通过平价联系；同一执行价、同一到期的看跌价格可以换算成看涨价格。若只看一边，可能漏掉另一边的报价信息；套利/密度检查要用整条链。

</details>

## 4. 曲面清洗与时间同步

### 4.1 三个时间戳

- 期权报价时间；
- 期货报价时间；
- 计算/快照时间。

三者不一致，隐波会被“旧期权价 + 新期货价”扭曲。

### 4.2 清洗规则

- 报价年龄超过阈值 → 标记 stale；
- 买卖价缺失/倒挂 → 剔除；
- 价格越界 → 剔除；
- 深度不足 → 降权/标记。

### 4.3 代码：报价年龄



```python

for age in [0, 30]:
    lower = max(F - K[0], 0)
    passed = age <= 5 and lower >= 0
    print("报价年龄", age, "分钟，通过时间/边界检查", passed)
```


**读数与应用**：年龄 30 分钟的报价不通过；真实曲面构建要对每条腿保留时间戳，不能用“最新快照”覆盖一切。

**常见错误**：

1. 用新期货价配旧期权价。
2. 只看中间价，不看买卖价。
3. 不记录清洗规则，结果无法复现。

**停下来想一想**：为什么“旧期权价 + 新期货价”会造出假隐波？

<details><summary>对照你的思路</summary>

期权价没变，但标的涨了；用更高的期货价反解，会把“标的变化”误当成“隐波变化”，造出很大的假隐波/假偏斜。时间同步是曲面构建的第一步。

</details>

## 5. 对冲手数为什么会变

### 5.1 三个原因

1. 标的动了，Delta 本身变（Gamma）；
2. 曲面动了，Vega × 斜率改变全导数；
3. 时间流逝，Delta/Gamma 变。

### 5.2 手数

若期权对应 100 单位标的，期货每手 10 单位：

$$\text{lots}=\frac{\Delta\times100}{10}.$$

**公式拆解**
- **符号**：$\Delta$ 期权 Delta（无量纲），100 为每张期权对应标的单位，10 为每手期货单位。
- **结构**：$lots=\Delta\times100/10$，即 Delta 暴露÷每手覆盖量；换成全导数得曲面移动手数。
- **代入**：$\Delta=0.3429$ → 3.429 手；$\Delta=0.3668$ → 3.668 手，差 0.239 手。
- **边界**：$\Delta=0$ 则 0 手；单位写错会差一个数量级。
- **用途**：对冲下单与手数敏感性；不是“对冲后无风险”，取整后仍有缺口。

固定 Delta 和移动 Delta 会给出不同手数；差异就是“曲面假设”的风险。

### 5.3 代码：两种手数



```python

units, mult = 100, 10
print("固定假设手数", round(fixed * units / mult, 3))
print("移动假设手数", round(moved * units / mult, 3))
print("差异", round((moved - fixed) * units / mult, 3))
```


**读数与应用**：差 0.1 手看似不大，但频繁调仓/大仓位会放大。对冲政策要写明用哪种曲面假设，并做敏感性。

**常见错误**：

1. 早盘/午盘用同一套 Delta。
2. 不更新曲面，只用昨天的隐波。
3. 对冲手数取整后不检查残余。

**停下来想一想**：为什么“固定行权价”假设在急涨急跌时容易失效？

<details><summary>对照你的思路</summary>

急涨急跌时，市场隐波/偏斜会快速变化；固定行权价假设认为某个 K 的隐波不变，但真实市场可能整体平移/旋转，导致对冲手数偏离。此时要用 sticky delta/动态曲面，或降低对冲频率、留缓冲。

</details>

| Sticky delta | 固定 $k=\ln(K/F)$ | 曲面跟随 |
| 全导数 | $V_F+V_\sigma\sigma_F$ | 真实 Delta |
| 隐含密度 | $e^{rT}\partial^2C/\partial K^2$ | 分布 |
| 时间同步 | 期权/期货/快照时间 | 清洗 |

### 八句话

1. 固定行权价和固定相对行权价，是两种曲面假设。
2. 真实 Delta 要加 Vega × 曲面斜率。
3. 曲率为负，密度为负，先查数据。
4. 旧期权价配新期货价，会造假隐波。
5. 报价年龄/买卖价/深度都要清洗。
6. 对冲手数随标的、曲面、时间变化。
7. 对冲政策要写明曲面假设。
8. 急涨急跌时，固定假设最容易失效。

## 本章习题

**1.** Sticky strike 和 sticky delta 的区别是什么？

**2.** 全导数公式中，第二项是什么？

**3.** 看涨价二阶差分为负，说明什么？

**4.** 为什么旧期权价 + 新期货价会造假隐波？

**5.** 期权对应 100 单位，Delta 0.5，期货每手 10 单位。对冲几手？

**6.** 固定 Delta 0.50、移动 Delta 0.55，手数差多少（100 单位/10）？

**7.** 报价年龄 30 分钟，应该怎么处理？

**8.** 为什么急涨急跌时固定行权价假设容易失效？

<details><summary>参考答案</summary>

**1.** Sticky strike 固定执行价的隐波；sticky delta 固定相对执行价 $k=\ln(K/F)$ 的隐波。

**2.** Vega × 隐波对标的的斜率。

**3.** 看涨价不凸/密度为负，可能有套利或数据问题。

**4.** 期权价未更新、期货价更新，反解把标的变化误当成隐波变化。

**5.** $0.5\times100/10=5$ 手。

**6.** $(0.55-0.50)\times100/10=0.5$ 手。

**7.** 标记 stale，降权/剔除，不用于曲面。

**8.** 隐波/偏斜快速变化，固定 K 的隐波假设与市场脱节。
