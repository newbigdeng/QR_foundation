## 1. 向量、内积、长度、距离

### 1.1 向量是什么

一个向量就是一列数：

$$a=(a_1,a_2,\dots,a_n).$$

在量化里，它可以是：

- 某一天所有资产的因子值（截面向量）；
- 某个资产过去 $n$ 天的收益（时间序列向量）；
- 一个组合的权重向量 $w=(w_1,\dots,w_n)$。

向量的长度（范数）：

$$||a||=\sqrt{a_1^2+\dots+a_n^2}.$$

两个向量的内积：

$$a\cdot b=\sum_{i=1}^n a_i b_i.$$

内积衡量“同方向的程度”：同号多，内积大；正负抵消，内积小。

### 1.2 夹角与相关

夹角余弦：

$$\cos\theta=\frac{a\cdot b}{||a||\,||b||}.$$

若 $a,b$ 已经中心化（减去均值），余弦就是相关系数。**相关系数就是两个中心化向量的夹角余弦。** 这句话把统计和几何连起来了。

距离：

$$||a-b||=\sqrt{\sum_i(a_i-b_i)^2}.$$

在因子聚类、相似资产筛选里，距离比相关更常用；但距离受单位影响，先标准化。

### 1.3 代码：向量运算



```python

import numpy as np

a = np.array([1.0, 2.0, 3.0])
b = np.array([2.0, 0.0, -1.0])

print("内积", a @ b)
print("a 的长度", round(np.linalg.norm(a), 4))
print("a、b 的距离", round(np.linalg.norm(a - b), 4))
print("夹角余弦", round(a @ b / (np.linalg.norm(a) * np.linalg.norm(b)), 4))

# 中心化后，余弦就是相关系数
x = np.array([1.0, 2.0, 3.0, 4.0])
y = np.array([2.0, 1.0, 4.0, 3.0])
xc, yc = x - x.mean(), y - y.mean()
print("中心化余弦", round(xc @ yc / (np.linalg.norm(xc) * np.linalg.norm(yc)), 4))
print("numpy 相关", round(np.corrcoef(x, y)[0, 1], 4))
```


**读数与应用**：中心化余弦和 `np.corrcoef` 应该一致。若不一致，通常是忘了中心化，或者把“距离”当成了“相关”。

**常见错误**：

1. 用原始向量算余弦，把“均值高低”当成“方向相似”。
2. 把距离和相关混用。两个因子距离近，不一定相关高；尺度不同，距离会被大数值主导。
3. 忘记标准化。成交量（手）和收益率（小数）放一起，距离几乎只反映成交量。

**停下来想一想**：两个因子的相关系数为 0，是否说明它们独立？

<details><summary>对照你的思路</summary>

不说明。相关系数只衡量线性关系；$Y=X^2$ 时，$X$ 与 $Y$ 明显依赖，但线性相关可能为 0。要判断独立，需要看联合分布、条件分布，或者至少画散点图。因子研究里，“线性不相关”不等于“没有共同信息”。

</details>

## 2. 矩阵乘法与协方差矩阵

### 2.1 矩阵乘法

矩阵 $A$ 是 $m\times n$，$B$ 是 $n\times p$，乘积 $C=AB$ 是 $m\times p$：

$$C_{ij}=\sum_{k=1}^n A_{ik}B_{kj}.$$

内侧维数必须相等。可以把 $AB$ 看成“对 $B$ 的每一列做一次 $A$ 的线性组合”，也可以看成“$A$ 的每一行与 $B$ 的每一列做内积”。

### 2.2 用矩阵表示组合

设 $R$ 是 $T\times N$ 的收益矩阵（每行一个交易日，每列一个资产），$w$ 是 $N\times1$ 的权重。组合收益：

$$r=Rw,$$

是一个 $T\times1$ 的向量。第 $t$ 天的组合收益，就是第 $t$ 行与 $w$ 的内积。

### 2.3 协方差矩阵

把 $R$ 的每一列减去自己的均值，得到中心化矩阵 $\tilde R$。样本协方差矩阵：

$$S=\frac{1}{T-1}\tilde R^T\tilde R.$$

$S$ 是 $N\times N$：

- 对角线 $S_{ii}$ 是资产 $i$ 的方差；
- 非对角线 $S_{ij}$ 是资产 $i,j$ 的协方差；
- 对称：$S_{ij}=S_{ji}$。

组合方差：

$$\operatorname{Var}(w^TR)=w^TSw.$$

**公式拆解**
- **符号**：$w$ 权重、$R$ 收益矩阵、$S$ 协方差矩阵；$w^TSw$ 是组合方差（小数$^2$）。
- **结构**：展开为对角项（各自方差）$+$ 两倍交叉项（协方差），重复信息在非对角。
- **代入**：$\sigma_1=20\%$、$\sigma_2=10\%$、$\rho=0.3$、$w=(0.6,0.4)$ → 方差 $0.01888$，$\sigma_p=13.74\%$。
- **边界**：$S$ 须半正定；奇异时会算出“零风险组合”，是账面套利。
- **用途**：组合方差、风险预算、优化目标；窗口估计的 $S$ 不等于真实风险。

这就是“矩阵乘法 + 二次型”在组合里的核心公式。

### 2.4 代码：从收益矩阵到协方差矩阵



```python

import numpy as np

rng = np.random.default_rng(601)
returns = rng.normal(0, 0.01, size=(250, 4))
returns[:, 1] = 0.8 * returns[:, 0] + 0.2 * returns[:, 1]  # 制造相关性

S = np.cov(returns, rowvar=False)
print("协方差矩阵形状", S.shape)
print(S.round(6))
print("对角线开根（各资产波动）", np.sqrt(np.diag(S)).round(4))

w = np.array([0.4, 0.3, 0.2, 0.1])
print("组合方差", round(w @ S @ w, 8), "组合波动", round(np.sqrt(w @ S @ w), 4))
```


**读数与应用**：协方差矩阵的对角线是方差，开根是波动率；非对角线是协方差。组合方差 $w^TSw$ 不是各资产方差的加权平均，因为还有交叉项。若两个资产负相关，交叉项为负，组合方差会小于加权平均。

**常见错误**：

1. 用 `np.cov(returns)` 默认 `rowvar=True`，把行当成变量。收益矩阵通常是“行是时间、列是资产”，要写 `rowvar=False`。
2. 样本协方差除以 $T$ 而不是 $T-1$。差别在样本小时明显。
3. 把协方差当相关系数。协方差有单位，数值大小不能直接比较。

**停下来想一想**：如果两个资产的收益完全负相关，组合方差能不能降到 0？

<details><summary>对照你的思路</summary>

可以，在特定权重下。设两个资产方差 $\sigma_1^2,\sigma_2^2$，相关系数 $-1$，取 $w_1=\sigma_2/(\sigma_1+\sigma_2)$、$w_2=\sigma_1/(\sigma_1+\sigma_2)$，组合方差为 0。但现实中完全负相关很少见，而且权重可能要求卖空；还要考虑成本和约束。

</details>

## 3. 协方差矩阵为什么必须是半正定的

### 3.1 二次型

对任意权重 $w$，组合方差

$$w^TSw\ge0.$$

方差不可能为负，所以 $S$ 必须是**半正定**的：所有 $w$ 都让二次型非负。

等价条件：$S$ 的所有特征值 $\ge0$。

### 3.2 半正定有什么用

- 优化器要求协方差矩阵半正定，否则“最小方差”可能无解或落到负方差；
- 样本协方差在 $T<N$（观测少于资产）时，会有零特征值，矩阵不可逆；
- 估计误差会让小特征值变成负数，导致求逆爆炸。

实务中常用收缩（shrinkage）、因子模型、岭正则化来修。

### 3.3 手算一个 $2\times2$

$$S=\begin{pmatrix}1&0.8\\0.8&1\end{pmatrix}.$$

特征值：

$$\lambda_1=1.8,\qquad \lambda_2=0.2.$$

都大于 0，半正定。特征向量：

$$v_1=\frac{1}{\sqrt2}(1,1),\qquad v_2=\frac{1}{\sqrt2}(1,-1).$$

第一根轴是“同向”方向，方差 1.8；第二根轴是“多空”方向，方差 0.2。**组合风险主要来自第一根轴。**

### 3.4 代码：特征值



```python

import numpy as np

S = np.array([[1.0, 0.8], [0.8, 1.0]])
vals, vecs = np.linalg.eigh(S)

print("特征值", vals)
print("特征向量（列）\n", vecs.round(3))
print("迹", np.trace(S), "特征值之和", vals.sum())
print("行列式", round(np.linalg.det(S), 3), "特征值之积", round(vals.prod(), 3))
```


**读数与应用**：`np.linalg.eigh` 专门用于对称矩阵，返回的特征值按升序排列；`eig` 不保证顺序，也可能返回复数。协方差矩阵对称，优先用 `eigh`。

**常见错误**：

1. 用 `eig` 处理对称矩阵，得到复数或乱序。
2. 把特征向量当成“资产权重”。特征向量是方向，不是组合权重；要变成权重还要归一化、处理符号。
3. 忽略小特征值。小特征值方向上的方差很小，但求逆后会放大估计误差。

**停下来想一想**：为什么 $T<N$ 时样本协方差矩阵一定不可逆？

<details><summary>对照你的思路</summary>

$S=\tilde R^T\tilde R/(T-1)$，$\tilde R$ 是 $T\times N$。矩阵的秩最多是 $\min(T-1,N)$。若 $T-1<N$，秩小于 $N$，至少有一个零特征值，所以不可逆。这就是“资产比观测多”时，协方差矩阵求逆会爆炸的原因。

</details>

## 4. 特征向量：风险的“轴”

### 4.1 定义

若矩阵 $S$ 作用在向量 $v$ 上，只改变长度、不改变方向：

$$Sv=\lambda v,$$

称 $v$ 是特征向量，$\lambda$ 是特征值。

对协方差矩阵，特征向量是“风险的主轴”，特征值是该方向上的方差。

### 4.2 几何直觉

把资产收益看成云团。协方差矩阵描述云团的形状：

- 特征值大：云团在这个方向拉得长；
- 特征值小：云团在这个方向扁；
- 特征向量：云团的长轴/短轴方向。

PCA 就是找这些轴。

### 4.3 手算

对 $S=\begin{pmatrix}1&0.8\\0.8&1\end{pmatrix}$：

- 特征值 $1.8,0.2$；
- 第一轴 $(1,1)/\sqrt2$：两个资产同向，方差 1.8；
- 第二轴 $(1,-1)/\sqrt2$：两个资产反向，方差 0.2。

第一轴解释了 $1.8/(1.8+0.2)=90\%$ 的总方差。也就是说，这两个资产的“共同风险”占九成，剩下的多空差异只占一成。

**这就是“重复信息”的几何版本**：两个因子看起来是两个，其实九成在同一个方向上。

### 4.4 代码：特征分解



```python

import numpy as np

S = np.array([[1.0, 0.8], [0.8, 1.0]])
vals, vecs = np.linalg.eigh(S)
order = np.argsort(vals)[::-1]
vals, vecs = vals[order], vecs[:, order]

print("从大到小特征值", vals.round(3))
print("解释比例", (vals / vals.sum()).round(3))
print("第一轴", vecs[:, 0].round(3))
print("第二轴", vecs[:, 1].round(3))
```


**读数与应用**：第一轴解释 90%，说明“同向风险”是主导；第二轴只有 10%，是“多空相对风险”。如果你用这两个资产做多空组合，组合风险主要不是来自“选股”，而是来自“共同暴露”。

**常见错误**：

1. 把特征值大小当成“因子好坏”。特征值大只说明方差大，不说明预测力。
2. 忽略特征向量的符号。特征向量整体乘 $-1$ 仍是同一个方向；解释时要固定符号。
3. 在没标准化的数据上做 PCA。量纲大的变量会主导第一主成分。

**停下来想一想**：如果两个资产的相关系数是 0.99，第一主成分解释多少？

<details><summary>对照你的思路</summary>

对 $2\times2$ 相关矩阵 $\begin{pmatrix}1&\rho\\\rho&1\end{pmatrix}$，特征值是 $1+\rho$ 和 $1-\rho$。$\rho=0.99$ 时，第一主成分解释 $(1.99)/(2)=99.5\%$。也就是说，两个资产几乎就是同一个风险；分散化几乎没有作用。

</details>

## 5. PCA：第一主成分在最大化什么

### 5.1 目标

给定中心化（并可标准化）的收益矩阵 $X$（$T\times N$），找一个单位向量 $v$，让投影 $Xv$ 的方差最大：

$$\max_v\ v^TSv\quad\text{s.t.}\quad v^Tv=1.$$

拉格朗日函数：

$$L=v^TSv-\lambda(v^Tv-1).$$

对 $v$ 求导，令零：

$$2Sv-2\lambda v=0\quad\Rightarrow\quad Sv=\lambda v.$$

**公式拆解**
- **符号**：$X$ 中心化数据、$S$ 协方差、$v$ 单位方向、$Xv$ 得分；$\lambda$ 是该方向方差。
- **结构**：最大化投影方差 $v^TSv$，约束 $v^Tv=1$ 防尺度爆炸，解出 $Sv=\lambda v$。
- **代入**：$S=\begin{pmatrix}1&0.8\\0.8&1\end{pmatrix}$ → $\lambda=1.8,0.2$，第一主成分解释 90%。
- **边界**：特征值重复则方向不唯一；$v$ 与 $-v$ 等价，正负号要钉死。
- **用途**：降维、正交化、风险提取；解释比例是方差，不是预测力。

所以最优方向就是最大特征值对应的特征向量。第二主成分在与第一主成分正交的方向里找，依次类推。

### 5.2 载荷、得分、解释比例

- **载荷**（loading）：特征向量 $v$，表示“这个主成分由哪些资产组成”；
- **得分**（score）：$Xv$，表示“每天这个主成分的值”；
- **解释比例**：$\lambda_i/\sum_j\lambda_j$。

### 5.3 标准化

PCA 对尺度敏感。成交量（手）和收益率（小数）放一起，第一主成分会被成交量主导。通常先对每个变量做 z-score：

$$z_{t,i}=\frac{x_{t,i}-\bar x_i}{s_i}.$$

标准化后，协方差矩阵就是相关矩阵，PCA 找的是“相关性结构”。

### 5.4 代码：PCA



```python

import numpy as np

rng = np.random.default_rng(602)
X = rng.normal(size=(300, 5))
X[:, 1] = 0.9 * X[:, 0] + 0.1 * X[:, 1]
X[:, 2] = 0.7 * X[:, 1] + 0.3 * X[:, 2]
X = (X - X.mean(axis=0)) / X.std(axis=0)

S = np.cov(X, rowvar=False)
vals, vecs = np.linalg.eigh(S)
order = np.argsort(vals)[::-1]
vals, vecs = vals[order], vecs[:, order]

print("特征值", vals.round(3))
print("解释比例", (vals / vals.sum()).round(3))
print("第一主成分载荷", vecs[:, 0].round(3))

scores = X @ vecs[:, :2]
print("前两主成分方差", scores.var(axis=0).round(3))
```


**读数与应用**：前两个主成分通常解释大部分方差。第一主成分的载荷如果全是同号，说明它是“市场/共同方向”；如果正负交替，说明它是“多空/相对方向”。这两种主成分的经济含义完全不同，不能只看解释比例。

**常见错误**：

1. 不标准化就做 PCA。量纲大的变量会主导。
2. 把主成分当成“因子”。主成分是统计构造，没有经济含义；要解释它，得看载荷和它与已知因子的相关。
3. 用全样本做 PCA，再回测。PCA 的均值和标准差也用了未来信息；要滚动估计。

**停下来想一想**：第一主成分解释 80%，是否说明“只有一个因子”？

<details><summary>对照你的思路</summary>

不一定。80% 说明“共同方向”占主导，但剩下的 20% 可能包含多个有经济意义的相对方向。PCA 是按方差排序的，不是按预测力排序；第二、第三主成分方差小，但可能正是可交易的多空信号。不要因为解释比例小就丢掉。

</details>

## 6. 正交化与回归残差：怎样剥离旧因子的信息

### 6.1 投影

要把向量 $y$ 中“与 $x$ 同方向”的部分剥掉，先做投影：

$$\hat y=bx,\qquad b=\frac{x^Ty}{x^Tx}.$$

残差：

$$e=y-bx.$$

残差与 $x$ 正交：

$$x^Te=x^Ty-b\,x^Tx=0.$$

**公式拆解**
- **符号**：$x$ 旧因子、$y$ 新因子（同一截面、已去均值）；$b=\frac{x^Ty}{x^Tx}$，$e=y-bx$。
- **结构**：$bx$ 是投影，$e$ 是残差；$x^Te=0$ 说明残差不含 $x$ 的线性信息。
- **代入**：$x=(-1.5,-0.5,0.5,1.5)$、$y=(-3,-1,0,4)$ → $b=2.2$，$e=(0.3,0.1,-1.1,0.7)$。
- **边界**：$x$ 全相同→$x^Tx=0$ 无定义；不去均值则残差均值不为 0。
- **用途**：中性化、增量测试；残差只是线性无关，不是纯 alpha。

这就是“把 $x$ 能解释的部分去掉，剩下的 $e$ 就是 $y$ 的增量信息”。

### 6.2 因子中性化

假设你有一个新因子 $y$，旧因子 $x$ 已经能解释一部分。用 $y$ 对 $x$ 回归，取残差作为“正交化后的新因子”：

$$y^\perp=y-\hat\beta x.$$

若 $x$ 是多列（多个旧因子），用多元回归，残差与整个 $x$ 空间正交。

### 6.3 顺序依赖

Gram-Schmidt 正交化是有顺序的：先剥离 $x_1$，再剥离 $x_2$，得到的残差依赖顺序。换一个顺序，结果不同。这不是 bug，是“你先定义谁是被解释的增量”。

实务中要写明：

- 先剥离哪些因子（市场、行业、规模）；
- 用哪个窗口估计 $\beta$；
- 残差是否再标准化。

### 6.4 $R^2$

$$R^2=1-\frac{\operatorname{Var}(e)}{\operatorname{Var}(y)}.$$

$R^2$ 高，说明旧因子已经解释了很多；新因子的增量小。但 $R^2$ 高不代表新因子没用：可能它只在尾部/特定状态有用。

### 6.5 代码：正交化



```python

import numpy as np

x = np.array([-1.0, 0.0, 1.0])
y = np.array([-1.0, -1.0, 2.0])
x = x - x.mean()
y = y - y.mean()

b = (x @ y) / (x @ x)
e = y - b * x

print("beta", round(b, 4))
print("残差", e.round(4))
print("x 与残差的内积", round(x @ e, 12))
print("R^2", round(1 - e.var() / y.var(), 4))
```


**读数与应用**：残差与 $x$ 的内积应该接近 0（浮点误差内）。$R^2$ 告诉你“旧因子解释了多少”，残差告诉你“还剩多少”。中性化后的因子，经济含义变成“在旧因子相同的情况下，新因子的相对高低”。

**常见错误**：

1. 把“正交”当成“独立”。正交只说明线性无关；非线性关系仍可能存在。
2. 用全样本估计 $\beta$，再用同一段算残差。这是偷看未来。
3. 不说明顺序。先剥离市场还是先剥离行业，结果不同；报告要写清。

**停下来想一想**：正交化后的因子，和原来的因子相关性是多少？

<details><summary>对照你的思路</summary>

和用来回归的那组 $x$ 线性相关为 0（在样本内）。但和 $y$ 本身不是一回事；正交化后的因子仍保留 $y$ 中未被 $x$ 解释的部分。若 $x$ 只能解释 $y$ 的 10%，残差和 $y$ 的相关会很高；若 $x$ 解释了 90%，残差就只剩很少的增量。

</details>

## 7. SVD：病态矩阵为什么放大噪声

### 7.1 分解

任意 $T\times N$ 矩阵可以写成

$$X=UDV^T,$$

- $U$：$T\times r$，正交列；
- $D$：$r\times r$，对角，元素是奇异值 $\sigma_1\ge\sigma_2\ge\dots\ge0$；
- $V$：$N\times r$，正交列；
- $r$ 是秩。

奇异值是“每个方向的伸缩倍数”。$\sigma_i$ 很小，说明这个方向几乎无法从数据里区分。

### 7.2 条件数与病态

条件数：

$$\kappa=\frac{\sigma_{\max}}{\sigma_{\min}}.$$

$\kappa$ 很大，矩阵接近奇异；求逆会放大噪声。比如两个因子几乎一样，协方差矩阵的最小特征值接近 0，求逆后权重会爆炸。

### 7.3 伪逆与岭回归

当矩阵不可逆时，用伪逆：

$$X^+=VD^{-1}U^T,$$

并把太小的 $\sigma_i$ 截断。或者加岭惩罚：

$$\hat\beta=(X^TX+\lambda I)^{-1}X^Ty.$$

$\lambda$ 越大，收缩越强，方差越小、偏差越大。

### 7.4 代码：奇异值与条件数



```python

import numpy as np

rng = np.random.default_rng(603)
X = rng.normal(size=(200, 3))
X[:, 2] = X[:, 0] + 0.001 * X[:, 2]   # 第三列几乎是第一列的复制

U, s, Vt = np.linalg.svd(X, full_matrices=False)
print("奇异值", s.round(4))
print("条件数", round(s[0] / s[-1], 1))
print("数值秩", np.linalg.matrix_rank(X))
```


**读数与应用**：第三列几乎等于第一列，最小奇异值很小，条件数很大。此时用 `np.linalg.inv(X.T@X)` 求回归系数，会放大舍入误差；应该用 `lstsq`、伪逆或岭回归。

**常见错误**：

1. 看到矩阵“可逆”就放心。数值上接近奇异，求逆仍然不稳定。
2. 用条件数判断“模型好不好”。条件数只说明数值稳定性，不说明预测力。
3. 岭回归不标准化。惩罚项对尺度敏感，先标准化再惩罚。

**停下来想一想**：条件数 1,000 是否意味着模型一定有问题？

<details><summary>对照你的思路</summary>

不一定。条件数大说明“输入方向很难区分”，参数估计的方差会大；但如果样本足够大、噪声足够小，预测仍可能稳定。要看参数的标准误、样本外表现和业务约束。条件数是警告，不是判决。

</details>

## 8. 多因子模型：收益和风险怎样分解

### 8.1 收益分解

设资产 $i$ 在 $t$ 期的收益：

$$r_{t,i}=\sum_{k=1}^K B_{i,k}f_{t,k}+\epsilon_{t,i}.$$

写成矩阵：

$$R=Bf+\epsilon.$$

- $B$：$N\times K$ 的暴露矩阵（每个资产对每个因子的敏感度）；
- $f$：$K\times1$ 的因子收益；
- $\epsilon$：个股/个券的特质收益。

### 8.2 两种回归

- **时间序列回归**：对每个资产，用历史 $r_{t,i}$ 对 $f_t$ 回归，估 $B_{i,\cdot}$；
- **截面回归**：在每个时点，用 $r_{t,i}$ 对 $B_{i,\cdot}$ 回归，估 $f_t$。

两种回归的行列含义不同，不能混。Fama–MacBeth 就是“先时序估暴露，再截面估因子收益，最后对因子收益做统计”。

### 8.3 风险分解

若因子收益协方差为 $\Sigma_f$，特质收益协方差为 $\Psi$（常假设对角），资产收益协方差：

$$S=B\Sigma_fB^T+\Psi.$$

组合风险：

$$\sigma_p^2=w^TSw=w^TB\Sigma_fB^Tw+w^T\Psi w.$$

第一项是“共同因子风险”，第二项是“特质风险”。

### 8.4 边际风险贡献

组合方差对权重 $w_i$ 的边际贡献：

$$\frac{\partial\sigma_p^2}{\partial w_i}=2(Sw)_i.$$

风险贡献占比：

$$RC_i=\frac{w_i(Sw)_i}{w^TSw},\qquad \sum_i RC_i=1.$$

**公式拆解**
- **符号**：$w_i$ 权重、$S$ 协方差；$(Sw)_i$ 是边际风险，$\sigma_p^2=w^TSw$ 是总方差。
- **结构**：导数 $2(Sw)_i$ 乘 $w_i$ 得方差贡献，再除以总方差成占比，占比和恒为 1。
- **代入**：沿用上例 → $RC_1=0.01584/0.01888=83.9\%$，$RC_2=16.1\%$。
- **边界**：$\sigma_p^2=0$ 无定义；负权重时 $RC<0$ 正常，不要截断。
- **用途**：风险预算、风险平价；风险贡献不等于收益贡献。

这告诉你“哪个仓位在主导风险”，而不是“哪个仓位最大”。

### 8.5 代码：估计暴露与风险分解



```python

import numpy as np

rng = np.random.default_rng(604)
n = 300
B = np.array([[1.0, 0.2], [0.8, -0.3], [0.5, 0.6], [1.2, 0.1]])
f = rng.normal(0, 0.01, size=(n, 2))
e = rng.normal(0, 0.005, size=(n, 4))
R = f @ B.T + e

B_hat = np.linalg.lstsq(f, R, rcond=None)[0].T
print("真实 B\n", B)
print("估计 B\n", B_hat.round(2))

Sigma_f = np.cov(f, rowvar=False)
Psi = np.diag(np.var(e, axis=0, ddof=1))
S_model = B @ Sigma_f @ B.T + Psi
print("模型协方差\n", S_model.round(6))

w = np.array([0.4, 0.3, 0.2, 0.1])
port_var = w @ S_model @ w
contrib = w * (S_model @ w)
print("组合波动", round(np.sqrt(port_var), 5))
print("风险贡献占比", (contrib / port_var).round(3))
```


**读数与应用**：估计的 $B$ 应该接近真实 $B$；模型协方差把共同因子风险和特质风险分开。风险贡献占比可能和权重不成比例：一个 40% 的仓位，可能贡献 60% 的风险，因为它和别的资产高度相关。

**常见错误**：

1. 用全样本估计 $B$，再回测同一段。暴露用了未来信息。
2. 忽略特质风险。只用 $B\Sigma_fB^T$ 会低估组合风险。
3. 把“权重最大”当成“风险最大”。要看边际贡献。

**停下来想一想**：两个资产权重各 50%，风险贡献一定各 50% 吗？

<details><summary>对照你的思路</summary>

不一定。若两个资产高度相关，且波动率不同，风险贡献会偏向波动大/相关性高的那个。极端情况下，一个 10% 的仓位可能贡献 50% 的风险。风险预算就是按贡献分配，而不是按权重分配。

</details>

## 9. 共线性与岭回归：因子太像了怎么办

### 9.1 共线性的信号

- 两个因子的相关系数接近 1；
- 回归系数很大、符号和直觉相反；
- 加/减一个样本，系数剧烈变化；
- 条件数很大，VIF 很高。

### 9.2 为什么 OLS 会不稳定

当 $X^TX$ 接近奇异时，它的最小特征值接近 0，$(X^TX)^{-1}$ 的特征值很大，误差被放大。系数估计的方差爆炸，预测反而不稳。

### 9.3 三种处理

1. **删因子**：简单，但可能删掉有增量的信息；
2. **正交化/合成**：把重复的因子合成一个，或保留残差；
3. **岭回归**：加惩罚 $\lambda||\beta||^2$，牺牲一点偏差，换方差下降。

岭回归的解：

$$\hat\beta_\lambda=(X^TX+\lambda I)^{-1}X^Ty.$$

$\lambda\to0$ 退化成 OLS；$\lambda\to\infty$ 系数趋于 0。用交叉验证选 $\lambda$。

### 9.4 代码：共线性与岭回归



```python

import numpy as np

rng = np.random.default_rng(605)
n = 100
x1 = rng.normal(size=n)
x2 = x1 + 0.0001 * rng.normal(size=n)
y = 2 * x1 + 3 * x2 + rng.normal(0, 0.1, size=n)

X = np.column_stack([np.ones(n), x1, x2])
beta_ols = np.linalg.lstsq(X, y, rcond=None)[0]
print("OLS", beta_ols.round(2))

for lam in [0.01, 0.1, 1.0]:
    penalty = lam * np.diag([0.0, 1.0, 1.0])   # 不惩罚截距
    beta_r = np.linalg.solve(X.T @ X + penalty, X.T @ y)
    print("ridge", lam, beta_r.round(2))
```


**读数与应用**：$x_1$ 与 $x_2$ 几乎完全共线，OLS 的系数可能很大、符号不稳；岭回归把系数往 0 拉，数值更稳，但代价是偏差。真实因子里，两个因子相关 0.95 很常见；不要只看 OLS 的 t 值，要看系数稳定性和样本外。

**常见错误**：

1. 看到共线就删因子。删掉的可能是有经济意义的那一个。
2. 岭回归不标准化。惩罚项对尺度敏感，先 z-score。
3. 用训练集的 $\lambda$ 调到测试集上。$\lambda$ 要用交叉验证/滚动验证选。

**停下来想一想**：共线性下，预测值一定不准吗？

<details><summary>对照你的思路</summary>

不一定。共线性主要让“单个系数的解释”不稳定，但预测值可能仍然稳，因为 $X\hat\beta$ 在共线方向上的组合是确定的。如果你只关心预测，不关心系数解释，共线性未必是致命问题；但如果你要用系数做归因/风控，就必须处理。

</details>

## 本章小结

### 一张公式表

| 概念 | 公式 | 用途 |
|---|---|---|
| 内积 | $a\cdot b=\sum a_ib_i$ | 方向相似度 |
| 范数 | $||a||=\sqrt{a\cdot a}$ | 长度/距离 |
| 协方差矩阵 | $S=\tilde R^T\tilde R/(T-1)$ | 资产共同波动 |
| 组合方差 | $w^TSw$ | 组合风险 |
| 特征分解 | $Sv=\lambda v$ | 风险主轴 |
| PCA | $\max v^TSv$ s.t. $v^Tv=1$ | 降维/共同因子 |
| 正交化 | $e=y-\frac{x^Ty}{x^Tx}x$ | 剥离旧因子 |
| SVD | $X=UDV^T$ | 秩/条件数/伪逆 |
| 多因子风险 | $S=B\Sigma_fB^T+\Psi$ | 风险分解 |
| 风险贡献 | $RC_i=w_i(Sw)_i/(w^TSw)$ | 风险预算 |
| 岭回归 | $(X^TX+\lambda I)^{-1}X^Ty$ | 共线性 |

### 八句话

1. 相关系数是中心化向量的夹角余弦。
2. 矩阵乘法是“线性组合”的批量写法。
3. 协方差矩阵必须半正定，否则组合方差会为负。
4. 特征向量是风险的主轴，特征值是那个方向的方差。
5. PCA 按方差排序，不按预测力排序。
6. 正交化剥掉的是线性信息，不是全部信息。
7. 条件数大，求逆会放大噪声；用 SVD/岭回归。
8. 多因子模型把风险拆成“共同因子”和“特质”两块。

**6.** 多因子模型 $S=B\Sigma_fB^T+\Psi$。若 $B$ 是 $4\times2$，$\Sigma_f$ 是 $2\times2$，$\Psi$ 是 $4\times4$，$S$ 的维度是多少？

**7.** 权重 $w=(0.5,0.5)$，$S=\begin{pmatrix}4&2\\2&9\end{pmatrix}$。求组合方差、两个资产的风险贡献占比。

**8.** 两个因子相关 0.98，OLS 系数很大且符号相反。你会怎么处理？至少说两种方法。

<details><summary>参考答案</summary>

**1.** 内积 $1\times2+2\times0+3\times(-1)=-1$；$||a||=\sqrt{14}\approx3.742$；距离 $||a-b||=||(-1,2,4)||=\sqrt{21}\approx4.583$；余弦 $-1/(\sqrt{14}\sqrt5)\approx-0.120$。

**2.** 特征值：$(4-\lambda)(9-\lambda)-4=0$，$\lambda^2-13\lambda+32=0$，$\lambda=(13\pm\sqrt{169-128})/2=(13\pm\sqrt{41})/2\approx8.70,4.30$。第一主成分解释 $8.70/13\approx66.9\%$。特征向量解 $(4-\lambda)v_1+2v_2=0$，归一化即可。

**3.** 总和 4.5；第一 $3/4.5=66.7\%$；第二 $1/4.5=22.2\%$。

**4.** $x^Ty=1-3=-2$，$x^Tx=2$，$b=-1$；残差 $e=y-(-1)x=(2,2,2)$；$x^Te=2-2=0$。

**5.** 输入方向几乎无法区分，$(X^TX)^{-1}$ 放大噪声；系数方差大、符号不稳，预测可能仍稳但解释不可靠。用 SVD/岭回归/正交化。

**6.** $B\Sigma_fB^T$ 是 $4\times4$；$\Psi$ 是 $4\times4$；$S$ 是 $4\times4$。

**7.** $Sw=(4\times0.5+2\times0.5,\ 2\times0.5+9\times0.5)=(3,5.5)$；组合方差 $w^TSw=0.5\times3+0.5\times5.5=4.25$；贡献 $w_i(Sw)_i=(1.5,2.75)$，占比 $1.5/4.25=35.3\%$、$2.75/4.25=64.7\%$。

**8.** 删掉一个/合成一个；做主成分或正交化；用岭回归；用先验/经济逻辑保留一个。核心是：不要直接解释 OLS 系数。
