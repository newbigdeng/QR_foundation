# 10 真实分钟实验：下破前的下行效率

> **这一章是一个上午的真实数据实验。** 目标不是“证明因子有效”，而是检查：你能否把 1 分钟数据正确合成 5 分钟线，定义事件，算标签，并承认样本太小。

### 学习目标

- 能把 1 分钟数据按交易节合成 5 分钟线。
- 能定义事件、因子和标签，并生成事件表。
- 能判断样本量是否足够，写清限制。
- 能区分“这个上午发生了什么”和“因子是否有效”。


### 先修

第 01、03、08 章；会 pandas 的 resample/rolling。

## 在量化研究中的位置

- **谁用**：因子研究员、实习生、数据实验。
- **输入**：1 分钟 OHLCV、交易节。
- **输出**：5 分钟线、事件表、标签、限制说明。
- **在流程的哪一步**：真实数据的最小实验。
- **失败模式**：跨午休拼接、缺 bar 当 0、用 2 个事件下结论。
- **最小工作流**：读数据 → 按交易节合成 5 分钟 → 定义事件/因子/标签 → 事件表 → 结论与限制。

---
## 本章导读

用 2026-11-02 上午的真实 1 分钟数据，走一遍“合成 5 分钟 → 定义事件 → 算标签 → 承认样本太小”的完整实验。

### 环境准备

本章只用 pandas/numpy，无需额外安装；真实数据在 `data/rb_morning_1102.csv`。

## 1. 任务

数据：`data/rb_morning_1102.csv`，RB2701 在 2026-11-02 上午的 1 分钟线。

问题：

> 09:27–09:46 有一段明显下跌，09:46 后反弹。若在“跌破前跌得顺”时做空，10 分钟内能延续吗？

**限制**：

- 只有一个上午，不能据此判断因子有效；
- 没有盘口，entry/exit 是分钟价格代理；
- 事件窗口不重叠。

---

## 2. 数据字典

| 字段 | 含义 |
|---|---|
| end | 分钟结束时间（YYYYMMDDHHMM） |
| open/high/low/close | 分钟 OHLC |
| volume | 成交量 |
| oi | 持仓量 |

**时间对齐**：`end` 是分钟结束时刻；09:01 的 bar 覆盖 09:00–09:01。合成 5 分钟线时，用右端标记，09:05 的线覆盖 09:01–09:05。

---

## 3. 从 1 分钟到 5 分钟

### 3.1 规则

- 按交易节分组，跨午休不拼接；
- 每 5 根 1 分钟合成一根 5 分钟；
- 只有满 5 根的窗口才算完整；
- 开盘取第一根 open，最高取 max，最低取 min，收盘取最后一根 close。

### 3.2 代码：读数据、合成



```python

import pandas as pd
import numpy as np

raw = pd.read_csv("data/rb_morning_1102.csv")
raw.index = pd.to_datetime(raw.pop("end").astype(str), format="%Y%m%d%H%M")
print("分钟数", len(raw), "首末", raw.index.min(), raw.index.max())
print(raw.head(3).round(2))
```


**读数**：135 根 1 分钟，09:01–11:30。先检查 index 唯一、递增；再合成。



```python

bars = []
for sid, block in raw.groupby((raw.index.to_series().diff() > pd.Timedelta(minutes=1)).cumsum()):
    bins = block.resample("5min", closed="right", label="right")
    five = bins.agg({"open": "first", "high": "max", "low": "min", "close": "last", "volume": "sum"})
    five = five[bins.size() == 5].copy()
    five["segment"] = sid
    bars.append(five)

candles = pd.concat(bars)
print("完整 5 分钟线", len(candles))
print(candles.head(3).round(2))
```


**读数**：只保留满 5 根的窗口；缺分钟的窗口被丢掉，不能当“价格没变”。

---

## 4. 事件、因子、标签

### 4.1 定义

- 事件：5 分钟收盘跌破前两根最低价；
- 因子：突破前 5 个收盘价的下行效率（公式逐项拆解见 01 §2.2）；
- 标签：下一分钟开盘入场，第 10 分钟收盘出场（做空：entry − exit）。

### 4.2 代码：事件表



```python

c = candles["close"]
candles["score"] = ((c.shift(4) - c) / c.diff().abs().rolling(4).sum().replace(0, np.nan)).shift(1)
candles["boundary"] = candles["low"].shift(1).rolling(2).min()
candles["breakout"] = c < candles["boundary"]

events = []
busy_until = None
for t, row in candles[candles["breakout"] & candles["score"].notna()].iterrows():
    idx = pd.date_range(t + pd.Timedelta(minutes=1), periods=10, freq="min")
    if not idx.isin(raw.index).all():
        continue
    path = raw.loc[idx]
    entry = path["open"].iloc[0]
    keep = busy_until is None or t >= busy_until
    if keep:
        busy_until = idx[-1]
    events.append({"time": t, "score": row["score"], "move": entry - path["close"].iloc[-1],
                   "best": entry - path["low"].min(), "worst": entry - path["high"].max()})

result = pd.DataFrame(events, columns=["time", "score", "move", "best", "worst"])
print(result.round({"score": 3, "move": 3, "best": 3, "worst": 3}).to_string(index=False))
```


**读数**：09:45 的事件效率 0.333，之后 10 分钟顺向 −5 点（逆向），最好只有 +2；11:20 的事件效率 0，顺向 −2。**按这个样本，效率高的一组反而没有延续。**

---

## 5. 结论与限制

- 09:45 那根跌破后，价格先下 2 点、再上 5 点，最后 −5 点；方向没延续；
- 11:20 的事件也亏；
- 只有 2 个事件、1 个上午，**不能判断因子有效/无效**；
- 需要多日样本、盘口、成本，才能做统计结论。

### 5.1 为什么 09:45 是反例

它满足“跌破 + 效率 0.333 + 有完整 10 分钟窗口”，看起来是合格事件；但后续路径是“先下 2 点、再上 5 点”。这说明：

- 跌破后的前 2 点可能是“假突破”；
- 只看“最终移动”会把“先给希望再反转”算成一次失败；
- 若用“先触线”标签，可能在 +2 时止盈、也可能在 −5 时止损，结果完全不同。

### 5.2 三种标签

| 标签 | 09:45 事件 | 回答的问题 |
|---|---|---|
| 10 分钟末移动 | −5 | 期末方向 |
| MFE/MAE | +2 / −5 | 途中最好/最坏 |
| 先触线（+2/−3） | 先 +2 还是先 −3？ | 路径先后 |

同一个事件，用不同标签会得到不同结论。**报告要写清标签，否则无法比较。**

### 5.3 如果要做多日

1. 把同一规则套到 20–60 个交易日；
2. 按交易日分组，报独立事件数；
3. 加盘口/成交价，算可执行净额；
4. 用“先触线”或“固定 10 分钟”两种标签做对照；
5. 样本外验证，不在这 2 个事件上调参。

这一课的价值是：**你会不会在只有 2 个样本时，忍住不下结论。**

---

## 本章小结

- 1 分钟 → 5 分钟要按交易节分组，缺 bar 不能补 0；
- 事件 = 跌破前两根最低价；因子 = 下行效率；标签 = 下一分钟开盘到第 10 分钟收盘；
- 09:45 事件效率 0.333，后续 −5 点，说明“跌得顺”不等于“会延续”；
- 样本只有 2 个事件、1 个上午，不能下统计结论。
