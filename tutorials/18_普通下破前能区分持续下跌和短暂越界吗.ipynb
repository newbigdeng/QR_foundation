## 2. 数据字典

数据文件：`data/rb_breakout_events.csv`，由 RB2701 历史分钟数据整理。

| 字段 | 含义 |
|---|---|
| day | 交易日（夜盘归到下一交易日） |
| signal | 信号时刻（最后一根完整 5 分钟线） |
| eff | 下行效率，突破前 5 个收盘价计算（公式见 01 §2.2） |
| floor | 突破前 10 分钟最低价 |
| width | 突破前 20 分钟高低差 |
| trigger | 触发价（跌破的边界） |
| entry | 下一分钟开盘价（做空参考入场） |
| c1–c10 | 后续 10 分钟的收盘价 |

**事件定义**：5 分钟收盘跌破前两根完整 5 分钟线的最低价。  
**标签**：入场价 − 第 10 分钟收盘价（做空方向，正数表示下跌）。  
**样本**：256 次事件、72 个交易日；数据已参与过研究，不是全新样本。

**重要限制**：

- 没有盘口，只有分钟价格；entry/c10 是价格代理，不是可成交回报；
- 事件窗口不重叠；
- 手续费、价差要另算。

### 2.1 三个时间尺度

这道题有三个时钟：

1. **信号时钟**：5 分钟线收盘，决定“事件成立”；
2. **执行时钟**：下一分钟开盘，决定 entry；
3. **标签时钟**：之后 10 分钟，决定 c1–c10。

三者不能混。若用信号那根 5 分钟线的收盘价当 entry，就用了“事件成立时已经过去”的价格；若用第 11 分钟的数据算标签，就多算了。

夜盘归属：21:45 的信号归到下一个交易日；跨午休/夜盘的窗口不拼接。样本只保留“有完整 10 分钟后续”的事件，这是事后限制，实盘下单时你不知道后面是否完整。

### 3.1 为什么用下行效率

交易员的原话是“走得越顺”。把它变成一个数，最直接的是“净跌幅 / 总路程”：净跌幅大、路程短，说明一路向下；净跌幅小、路程长，说明来回震荡。

这个定义有三个好处：可复现、无量纲、方向明确（越高越顺）。坏处是：它只看突破前，不看突破后的成交量/盘口；分母为 0 时要返回缺失。

## 5. 参考解法

### 5.1 先看数据



```python

import pandas as pd

df = pd.read_csv("data/rb_breakout_events.csv", parse_dates=["day", "signal"])
print("事件数", len(df), "交易日", df["day"].nunique())
print(df.head(3).to_string())
```


### 5.2 算标签

做空方向：入场价 − 后续收盘价。正数表示下跌（顺向），负数表示上涨（逆向）。



```python

price_cols = [f"c{i}" for i in range(1, 11)]
df["move"] = df["entry"] - df["c10"]
df["mfe"] = df["entry"] - df[price_cols].min(axis=1)
df["mae"] = df["entry"] - df[price_cols].max(axis=1)

print(df[["move", "mfe", "mae"]].describe().round(3))
```


**读数**：平均顺向变化 −0.36 点，说明“普通下破”整体并不是一个赚钱的做空信号；MFE 平均 1.28 点，MAE 平均 −1.80 点，逆向波动比顺向空间大。

### 5.3 训练段定阈值，测试段只应用



```python

train = df[df["day"] <= pd.Timestamp("2026-08-31")]
test = df[df["day"] > pd.Timestamp("2026-08-31")]
lo, hi = train["eff"].quantile([1 / 3, 2 / 3])

def group_table(d):
    d = d.copy()
    d["group"] = pd.cut(d["eff"], [-float("inf"), lo, hi, float("inf")],
                        labels=["low", "mid", "high"])
    return d.groupby("group", observed=True).agg(
        n=("move", "size"), days=("day", "nunique"),
        mean=("move", "mean"), median=("move", "median"))

print("训练段阈值", round(lo, 3), round(hi, 3))
print("训练段")
print(group_table(train).round(3))
print("测试段")
print(group_table(test).round(3))
```


**读数**：训练段阈值 −0.20 / 0.33。测试段高分组平均 −0.15 点，低分组 −1.00 点；高分组少亏，但两组都是负的。

### 5.4 为什么在训练段定阈值

阈值（−0.20/0.33）只用 2026-08-31 之前的数据算；测试段原样应用，不重新分位。若用全样本分位，测试段的“高分组”就用了未来信息，回测会虚高。

这是“研究”和“回测”的分界：研究可以在训练段反复试；回测/前向只能应用。

### 5.5 同日配对与 bootstrap



```python

import numpy as np

test = test.copy()
test["group"] = pd.cut(test["eff"], [-float("inf"), lo, hi, float("inf")],
                       labels=["low", "mid", "high"])
pivot = test.pivot_table(index="day", columns="group", values="move",
                         aggfunc="mean", observed=True)
paired = (pivot["high"] - pivot["low"]).dropna()

rng = np.random.default_rng(1)
boot = rng.choice(paired.to_numpy(), size=(5000, len(paired)), replace=True).mean(axis=1)

print("配对交易日", len(paired), "高减低均值", round(paired.mean(), 3))
print("95% bootstrap 区间", tuple(round(float(v), 3) for v in np.quantile(boot, [0.025, 0.975])))
```


**读数**：14 个配对日，高减低 +0.70 点；95% 区间 [−0.98, 2.68]，跨过 0。**方向对，但不显著。**

### 5.6 为什么要同日配对

不同交易日的行情差异很大；若高分组恰好集中在某几天，直接比较高/低组会把“日期效应”当成“因子效应”。同日配对只比较“同一天内高组和低组的差”，再把日差做 bootstrap。

代价是：只保留同时有高、低组的交易日，样本从 104 个事件降到 14 个配对日；结论的不确定性更大。

### 5.7 成本账



```python

for cost in [2, 4]:
    high_net = (test.loc[test["group"] == "high", "move"].mean() - cost) * 10 - 6
    low_net = (test.loc[test["group"] == "low", "move"].mean() - cost) * 10 - 6
    print("每边损耗", cost, "点：高分组净利", round(high_net, 2), "低分组净利", round(low_net, 2))
```


**读数**：即使只按 2 点价差 + 6 元手续费，高分组每笔净亏约 27.5 元，低分组亏 36 元。**没有可交易的正净额。**

### 5.8 研究决定

结论要写成“能被反驳的一句话”：

> 在这批 RB2701 事件里，下行效率高的组比低的组少亏约 0.7 点，但 95% 区间跨 0，且扣掉 2 点价差 + 6 元手续费后两组都亏。**现有证据不足以支持“下破前越顺、下破后越容易延续”的可交易版本。**

下一步：

1. 扩样本到更多独立交易日；
2. 加盘口/成交数据，算可成交净额；
3. 检查“高分组少亏”是否只是少数交易日；
4. 如果仍想继续，先做前向观察，不要直接上线。

- 用“平均赚 0.7 点”直接说“策略有效”：−10。

### 6.1 怎么用这张评分表

先自己打分，再看参考解法。重点不是“结论对不对”，而是：计划里的定义别人能不能复算、对照是否公平、成本是否算到一手、结论是否承认不确定性。

如果计划写“用效率分组”，但没写阈值哪段定、缺失怎么处理，这一项最多拿一半。

---

## 7. 常见错误

1. **先看答案再写计划**。计划会不自觉地围绕结果编。
2. **用全样本分位数**。阈值用了未来信息。
3. **只看高分组**。没有“全做/被拒”，不知道筛子有没有用。
4. **用中间价算净利**。分钟收盘价不是可成交价。
5. **把 0.7 点当利润**。乘数、双边成本、样本量都要算。
6. **样本太少还下强结论**。14 个配对日、区间跨 0，只能算“初步”。

### 7.1 一个反例：把“少亏”当“有效”

高分组平均 −0.15 点、低分组 −1.00 点，看起来“高分组少亏 0.85 点”。但两组都是负的；扣 2 点价差后高分组每笔亏 27.5 元；14 个配对日、区间跨 0。

“相对少亏”不是“绝对赚钱”。报告里要把这两句话分开写。

---

## 本章小结

这道题的价值不在“找到因子”，而在走完一次**可能推翻自己**的研究：

- 先写计划，再碰数据；
- 训练段定阈值，测试段只应用；
- 全做/筛过/被拒三行账；
- 成本算到一手净利；
- 用区间/独立交易日承认不确定性；
- 结论要能被反例推翻。

**如果结论是“不可交易”，这也是一个有效的研究结论。**
