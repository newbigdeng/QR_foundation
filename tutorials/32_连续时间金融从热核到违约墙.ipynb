## 1. 扩散与热核：密度怎样演化

### 1.1 GBM 的终值密度

风险中性下：

$$\ln F_T\sim N\left(\ln F_0-\frac12\sigma^2T,\ \sigma^2T\right).$$

密度：

$$f(x)=\frac{1}{x\sigma\sqrt{2\pi T}}\exp\left(-\frac{(\ln x-\mu)^2}{2\sigma^2T}\right),$$

其中 $\mu=\ln F_0-\frac12\sigma^2T$。

**公式拆解**
- **符号**：$F_0$ 现价，$\sigma$ 波动率，$T$ 年，$\mu=\ln F_0-\frac12\sigma^2T$。
- **结构**：$f(x)=\frac{1}{x\sigma\sqrt{2\pi T}}\exp[-\frac{(\ln x-\mu)^2}{2\sigma^2T}]$，对数正态。
- **代入**：$F_0=100,\sigma=0.25,T=0.5$：$\mu=4.59$，$f(100)\approx0.0225$。
- **边界**：$x\to0$ 或 $\infty$ 时 $f\to0$。
- **用途**：把定价写成密度积分。

### 1.2 定价是积分

$$C=e^{-rT}\int_0^\infty \max(x-K,0)f(x)\,dx.$$

“热核”就是扩散方程的基本解；密度随时间的演化，就是热传导。

**公式拆解**
- **符号**：$C$ 看涨现值，$x$ 到期价，$K$ 执行价，$f$ 风险中性密度，$r,T$ 为贴现。
- **结构**：$C=e^{-rT}\int_K^\infty(x-K)f(x)dx$。
- **代入**：$F_0=K=100,T=0.5,\sigma=0.25,r=2\%$：$\int xf=53.52$、$\int Kf=46.48$，差 7.04，贴现后 6.97。
- **边界**：$K\to0$ 时 $C\to e^{-rT}F_0$；$T\to0$ 时退化为 $\max(F_0-K,0)$。
- **用途**：统一“定价=求期望”；注意积分限。

### 1.3 代码：积分 vs Black–76



```python

import numpy as np
from scipy.stats import norm
from scipy.integrate import quad

F0, K, T, vol, r = 100.0, 100.0, 0.5, 0.25, 0.02
mu = np.log(F0) - 0.5 * vol ** 2 * T
sigma = vol * np.sqrt(T)

def density(x):
    return norm.pdf((np.log(x) - mu) / sigma) / (x * sigma)

integral = quad(lambda x: max(x - K, 0) * density(x), K, 500)[0]
print("积分价", round(np.exp(-r * T) * integral, 4))
```


**读数与应用**：积分价应接近 Black–76 的 6.97。这就是“风险中性期望 × 贴现”。

**常见错误**：

1. 用真实漂移 $\mu$ 而不是 $r$。
2. 忘记 $-\frac12\sigma^2$。
3. 积分下限写成 0，支付为负。

**停下来想一想**：为什么密度积分和 Black–76 公式是同一件事？

<details><summary>对照你的思路</summary>

Black–76 是“对数正态密度 × 看涨支付”的解析积分；积分形式是定义，公式是结果。两者在同样的假设下等价；数值积分可以用来校验公式/代码。

</details>

## 2. 路径依赖：障碍/违约墙

### 2.1 只看终点不够

障碍期权在路径碰到障碍时生效/失效；只看 $F_T$ 会漏掉“中途触碰”。离散监控（每天/每分钟）和连续监控也不同。

### 2.2 蒙特卡洛

模拟很多条路径，逐条判断是否触碰障碍，再算支付平均。

### 2.3 代码：障碍概率



```python

rng = np.random.default_rng(2001)
n_paths, n_steps = 20_000, 100
dt = T / n_steps
z = rng.normal(size=(n_paths, n_steps))
paths = F0 * np.exp(np.cumsum(-0.5 * vol ** 2 * dt + vol * np.sqrt(dt) * z, axis=1))

barrier = 90.0
touch = (paths.min(axis=1) <= barrier).mean()
print("触碰障碍概率（离散监控）", round(float(touch), 4))
```


**读数与应用**：离散监控下，触碰概率取决于步数；步数越细，越接近连续监控。真实障碍条款要写清“收盘价/最高最低价/连续”。

**常见错误**：

1. 只用终点判断障碍。
2. 用日线高低点冒充连续监控。
3. 忽略跳空/停牌。

**停下来想一想**：为什么“离散监控”的障碍概率通常低于“连续监控”？

<details><summary>对照你的思路</summary>

离散监控只在某些时点检查，路径可能在两次检查之间穿过障碍又回来，没被记录；连续监控会抓到。步数越细，离散越接近连续。

</details>

## 3. 蒙特卡洛误差与方差缩减

### 3.1 标准误

若每条路径支付 $X_i$，均值 $\bar X$ 的标准误：

$$SE=\frac{s}{\sqrt n}.$$

95% 区间约 $\bar X\pm1.96SE$。

**公式拆解**
- **符号**：$n$ 路径数，$\bar X$ 均值，$s$ 为支付样本标准差，$SE$ 为标准误。
- **结构**：$SE=s/\sqrt n$；95% 区间约 $\bar X\pm1.96SE$。
- **代入**：$n=20000$、$SE=0.0817$：$s=0.0817\times141.42\approx11.55$，区间约 $\pm0.160$。
- **边界**：$n=1$ 时 $s$ 无定义；$n\to\infty$ 时 $SE\to0$ 但模型误差不消失。
- **用途**：报告 MC 区间、规划路径数；$SE$ 不是定价误差。

### 3.2 对偶变量

用 $Z$ 和 $-Z$ 成对模拟，支付平均；若支付关于 $Z$ 单调，方差会下降。

### 3.3 布朗桥

只在粗网格模拟，再用布朗桥补中间点，估计“路径是否触碰障碍”；比细网格便宜。

### 3.4 代码：标准误与对偶



```python

def payoff_from_z(z):
    path = F0 * np.exp(np.cumsum(-0.5 * vol ** 2 * dt + vol * np.sqrt(dt) * z))
    return max(path[-1] - K, 0)

z1 = rng.normal(size=(20_000, n_steps))
plain = np.array([payoff_from_z(z) for z in z1])
anti = 0.5 * (plain + np.array([payoff_from_z(-z) for z in z1]))

print("普通均值/标准误", round(plain.mean(), 4), round(plain.std(ddof=1) / np.sqrt(len(plain)), 4))
print("对偶均值/标准误", round(anti.mean(), 4), round(anti.std(ddof=1) / np.sqrt(len(anti)), 4))
```


**读数与应用**：对偶变量的标准误通常更小；均值应接近。真实定价要报告区间，不只报一个数。

**常见错误**：

1. 只报均值，不报标准误。
2. 对偶变量用错配对（不是 $Z$ 与 $-Z$）。
3. 路径数太少，区间很宽。

**停下来想一想**：为什么“对偶”能降方差？

<details><summary>对照你的思路</summary>

$Z$ 与 $-Z$ 的支付通常负相关：一个高、一个低；平均后波动下降。若支付关于 $Z$ 单调，负相关最强，方差下降最多。

</details>

## 4. Feynman–Kac：PDE 与期望的桥

### 4.1 结论

在风险中性下，期权价格既是 PDE 的解，也是贴现期望：

$$V(F,t)=E^Q\left[e^{-r(T-t)}\text{payoff}(F_T)\mid F_t=F\right].$$

这就是 Feynman–Kac：PDE 的初值问题 = 期望。

**公式拆解**
- **符号**：$V(F,t)$ 期权价，$E^Q$ 为风险中性条件期望，$r$ 利率，$T$ 到期。
- **结构**：$V(F,t)=E^Q[e^{-r(T-t)}\text{payoff}(F_T)\mid F_t=F]$，PDE 初值问题等于期望。
- **代入**：$F=K=100,T=0.5,\sigma=0.25$：未贴现期望 7.04，贴现后 6.97，与 Black–76 一致。
- **边界**：$t=T$ 时 $V=$payoff；美式是变分不等式，普通期望不成立。
- **用途**：解析、树、MC 互相校验；期望必须用 $Q$，不是真实漂移。

### 4.2 用途

- 解析公式难时，用蒙特卡洛算期望；
- 路径依赖/美式时，用 PDE/树/最小二乘；
- 两者可以互相校验。

**常见错误**：

1. 蒙特卡洛用真实漂移。
2. PDE 边界写错。
3. 不比较两种方法，无法发现 bug。

**停下来想一想**：为什么美式期权不能直接用普通蒙特卡洛？

<details><summary>对照你的思路</summary>

美式要在路径上做“继续持有 vs 立即行权”的最优决策；普通蒙特卡洛只能算到期支付，不能回去做最优停止。要用树/PDE/最小二乘蒙特卡洛（LSM）。
