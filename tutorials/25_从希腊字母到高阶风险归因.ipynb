## 1. 五个一阶希腊字母

### 1.1 定义

| 希腊字母 | 定义 | 单位 | 看涨符号 |
|---|---|---|---|
| Delta | $\partial V/\partial F$ | 元/点 | 正 |
| Gamma | $\partial^2 V/\partial F^2$ | Delta/点 | 正 |
| Vega | $\partial V/\partial\sigma$ | 元/波动率单位 | 正 |
| Theta | $\partial V/\partial t$ | 元/天 | 通常负 |
| Rho | $\partial V/\partial r$ | 元/利率单位 | 正 |

**单位最重要。** 同样是“Vega 20”，有的按“1 个波动率点（0.01）”报，有的按“1.0 波动率”报，差 100 倍。

### 1.2 Black–76 的解析式

看涨：

$$C=e^{-rT}[F\Phi(d_1)-K\Phi(d_2)].$$

$$\Delta=e^{-rT}\Phi(d_1),$$

$$\Gamma=\frac{e^{-rT}\phi(d_1)}{F\sigma\sqrt T},$$

$$\text{Vega}=e^{-rT}F\phi(d_1)\sqrt T.$$

**公式拆解**
- **符号**：$\Delta,\Gamma$、Vega 为偏导。
- **结构**：$\Delta=e^{-rT}\Phi(d_1)$，$\Gamma=e^{-rT}\phi(d_1)/(F\sigma\sqrt T)$，Vega$=e^{-rT}F\phi(d_1)\sqrt T$。
- **代入**：$F=K=100,T=0.5,\sigma=0.25$：$\Delta=0.53,\Gamma=0.022$、Vega$=27.8$。
- **边界**：深实值 $\Delta\to e^{-rT}$，虚值趋 0。
- **用途**：算风险表；$\Delta$≠概率。

Theta 的解析式更长，实务中常用有限差分；本章会用差分校验。

### 1.3 代码：定价与差分



```python

import numpy as np
from scipy.special import ndtr

def black76(F, K, T, vol, r=0.02, kind="call"):
    if T == 0:
        return max(F - K, 0) if kind == "call" else max(K - F, 0)
    w = vol * np.sqrt(T)
    d1 = (np.log(F / K) + 0.5 * w * w) / w
    d2 = d1 - w
    if kind == "call":
        return np.exp(-r * T) * (F * ndtr(d1) - K * ndtr(d2))
    return np.exp(-r * T) * (K * ndtr(-d2) - F * ndtr(-d1))

F, K, T, vol, r = 100.0, 100.0, 0.5, 0.25, 0.02
base = black76(F, K, T, vol, r)
print("看涨价格", round(base, 4))
```


**读数与应用**：平值看涨约 6.97 元。下面所有希腊字母都以这个价格为基准。

**常见错误**：

1. 把 Vega 的“1 个点”当成 1.0 波动率。1 个点通常是 0.01。
2. 把 Theta 的“每天”与“每年”混用。
3. 用现货价代替期货价。

**停下来想一想**：为什么 Gamma 和 Vega 在平值附近最大？

<details><summary>对照你的思路</summary>

平值时，期权最“犹豫”：标的小动就可能从实值变虚值，Delta 变化最快（Gamma 大）；波动率上升对平值期权的时间价值影响最大（Vega 大）。深度实值/虚值时，Delta 接近 0 或 1，变化慢，Gamma/Vega 小。

</details>

## 2. Delta、Gamma、Vega、Theta 的符号与量级

### 2.1 手算

$F=100,K=100,T=0.5,\sigma=0.25,r=0.02$，用差分：

- Delta $\approx0.53$；
- Gamma $\approx0.022$；
- Vega（每 1.0 波动率）$\approx27.8$；
- Theta（每年）$\approx-6.8$。

含义：

- 标的涨 1 点，期权约涨 0.53 元；
- 标的涨 1 点，Delta 约增加 0.022；
- 波动率涨 0.01（1 个点），期权约涨 0.278 元；
- 每过一天，期权约亏 0.019 元（$-6.8/365$）。

### 2.2 代码：四个希腊字母



```python

h, b, dt = 0.01, 1e-4, 1 / 365

delta = (black76(F + h, K, T, vol, r) - black76(F - h, K, T, vol, r)) / (2 * h)
gamma = (black76(F + h, K, T, vol, r) - 2 * base + black76(F - h, K, T, vol, r)) / h ** 2
vega = (black76(F, K, T, vol + b, r) - black76(F, K, T, vol - b, r)) / (2 * b)
theta = (black76(F, K, T - dt, vol, r) - base) / dt

print("Delta", round(delta, 4))
print("Gamma", round(gamma, 4))
print("Vega（每 1.0 波动率）", round(vega, 4), "每 1 个点", round(vega / 100, 4))
print("Theta（每年）", round(theta, 4), "每天", round(theta / 365, 4))
```


**读数与应用**：Delta 0.53、Gamma 0.022、Vega 27.8、Theta −6.8/年。报告里要写清“每 1.0 波动率”还是“每 1 个点”。

**常见错误**：

1. 把 Theta 的负号丢掉。
2. 把 Gamma 的单位写成“元”。它是 Delta/点。
3. 用有限差分时步长太大/太小；$h=0.01$、$b=10^{-4}$ 是常用起点。

**停下来想一想**：为什么 Delta 对冲后，Theta 仍然存在？

<details><summary>对照你的思路</summary>

Delta 对冲消掉一阶方向风险，但时间流逝（Theta）不是方向风险；只要持有期权，时间价值就在变化。对冲组合的损益仍然包含 Theta、Gamma、Vega。Delta 中性不等于“不亏不赚”。

</details>

## 3. 高阶希腊字母：Vanna、Volga、Charm

### 3.1 定义

- **Vanna**：$\partial^2V/\partial F\partial\sigma$，Delta 对波动率的敏感度，也是 Vega 对标的的敏感度；
- **Volga**：$\partial^2V/\partial\sigma^2$，Vega 对波动率的敏感度；
- **Charm**：$\partial^2V/\partial F\partial t$，Delta 随时间的变化。

### 3.2 为什么重要

当 Delta、Vega 都对冲掉后，标的和波动率“同时动”或“波动率大幅动”时，一阶项不够，Vanna/Volga 决定残余损益。

### 3.3 交叉差分

$$\text{Vanna}\approx\frac{V(F+h,\sigma+b)-V(F+h,\sigma-b)-V(F-h,\sigma+b)+V(F-h,\sigma-b)}{4hb}.$$

$$\text{Volga}\approx\frac{V(\sigma+b)-2V(\sigma)+V(\sigma-b)}{b^2}.$$

**公式拆解**
- **符号**：Vanna、Volga 为交叉与二阶偏导。
- **结构**：Vanna 为 $F,\sigma$ 交叉差分除以 $4hb$；Volga 为 $\sigma$ 二阶差分除以 $b^2$。
- **代入**：$\sigma$ 0.24→0.26 时 $\Delta$ 0.5285→0.5313，Vanna$\approx0.14$；Volga$\approx-0.87$。
- **边界**：$b\to0$ 时差分被舍入误差吞掉。
- **用途**：解释 Delta+Vega 后的残余。

### 3.4 代码：高阶希腊字母



```python

vanna = (
    black76(F + h, K, T, vol + b, r) - black76(F + h, K, T, vol - b, r)
    - black76(F - h, K, T, vol + b, r) + black76(F - h, K, T, vol - b, r)
) / (4 * h * b)

volga = (black76(F, K, T, vol + b, r) - 2 * base + black76(F, K, T, vol - b, r)) / b ** 2

print("Vanna", round(vanna, 6))
print("Volga", round(volga, 4))
```


**读数与应用**：本例 Vanna 为正、Volga 为负；符号会随期限、行权价、利率变化，不要背符号。两者都是“二阶交叉/二阶波动率”项，量级比 Delta/Vega 小，但在大变动时不能忽略。

**常见错误**：

1. 把 Vanna 当成“Delta 的 Delta”。
2. 不写单位；Vanna 的单位是“Delta/波动率单位”。
3. 用一阶近似处理“标的和波动率同时大动”。

**停下来想一想**：为什么“标的涨 + 波动率涨”时，只用 Delta + Vega 会低估/高估？

<details><summary>对照你的思路</summary>

Delta 和 Vega 是一阶项；两者同时动时，交叉项 Vanna 会贡献 $Vanna\,\Delta F\,\Delta\sigma$。若 Vanna 为正、两者同向，实际收益高于一阶近似；若符号相反，则低于。二阶项在大变动时不可忽略。

</details>

## 4. 组合风险表：从“每单位”到“现金希腊字母”

### 4.1 三步

1. 算每个合约的“每单位希腊字母”；
2. 乘手数、乘合约乘数；
3. 按合约加总。

$$\text{现金 Delta}_i=\Delta_i\times\text{lots}_i\times\text{mult}.$$

Gamma、Vega、Theta 同理。**不乘乘数，风险表会小看 10 倍/100 倍。**

**公式拆解**
- **符号**：$\Delta_i$、Vega$_i$ 为每单位 Greeks，lots 手数，mult 乘数。
- **结构**：现金 Delta$=\Delta\times\text{lots}\times\text{mult}$，Gamma/Vega/Theta 同构。
- **代入**：买 2 手 C100（$\Delta=0.53$）$=10.6$，卖 3 手 C105$=-12.6$，买 1 手 P95$=-3.5$，合计 $-5.5$。
- **边界**：mult$=1$ 时每单位即每手；Delta 为 0≠无风险。
- **用途**：统一金额单位再算对冲。

### 4.2 符号

- 买入看涨：Delta 正、Gamma 正、Vega 正、Theta 负；
- 卖出看涨：全部取反；
- 买入看跌：Delta 负、Gamma 正、Vega 正、Theta 负。

### 4.3 代码：三腿组合



```python

import pandas as pd

positions = [("C100", "call", 100, 2), ("C105", "call", 105, -3), ("P95", "put", 95, 1)]
mult = 10
rows = []

for name, kind, strike, lots in positions:
    price = black76(F, strike, T, vol, r, kind)
    d = (black76(F + h, strike, T, vol, r, kind) - black76(F - h, strike, T, vol, r, kind)) / (2 * h)
    g = (black76(F + h, strike, T, vol, r, kind) - 2 * price + black76(F - h, strike, T, vol, r, kind)) / h ** 2
    v = (black76(F, strike, T, vol + b, r, kind) - black76(F, strike, T, vol - b, r, kind)) / (2 * b)
    th = (black76(F, strike, T - dt, vol, r, kind) - price) / dt
    rows.append({"合约": name, "手数": lots, "价格": price,
                 "现金Delta": d * lots * mult, "现金Gamma": g * lots * mult,
                 "现金Vega": v * lots * mult, "现金Theta": th * lots * mult})

risk = pd.DataFrame(rows)
print(risk.round(3).to_string(index=False))
print("合计", risk[["现金Delta", "现金Gamma", "现金Vega", "现金Theta"]].sum().round(3).to_dict())
```


**读数与应用**：现金 Delta/Gamma/Vega/Theta 是“整个组合、按 1 点/1 波动率/1 年”的暴露。报告里要写单位，不能只写“Delta 2.3”。

**常见错误**：

1. 忘记乘合约乘数。
2. 卖出的符号没取反。
3. 把不同到期/不同标的的希腊字母直接相加。

**停下来想一想**：为什么不同到期的 Vega 不能直接相加？

<details><summary>对照你的思路</summary>

不同到期的期权对“哪个期限的波动率”敏感；若波动率曲面不是平行移动，短端和长端的 Vega 不能用一个数相加。实务中要按期限分桶（Vega bucketing），再做情景。

</details>

## 5. 情景归因：精确重估 vs 希腊字母近似

### 5.1 精确重估

把所有输入改成新值，重新定价：

$$\Delta V_{\text{exact}}=V(F+\Delta F,\sigma+\Delta\sigma,T-\Delta t)-V(F,\sigma,T).$$

### 5.2 希腊字母近似

$$\Delta V\approx\Delta\,\delta F+\frac12\Gamma(\delta F)^2+\text{Vega}\,\delta\sigma+\Theta\,\delta t.$$

### 5.3 残差

$$\text{残差}=\Delta V_{\text{exact}}-\Delta V_{\text{approx}}.$$

残差来自高阶项（Vanna、Volga、三阶）、离散误差、模型假设。残差大，说明一阶/二阶近似不够。

**公式拆解**
- **符号**：$\Delta V_{\text{exact}}$ 精确重估，$\Delta V_{\text{approx}}$ 近似。
- **结构**：残差 $=$ 精确 $-$ 近似，来自 Vanna/Volga/三阶与离散误差。
- **代入**：三腿组合 $\delta F=1,\delta\sigma=0.01$、1 天：精确 $-5.78$，近似 $-5.60$，残差 $-0.18$。
- **边界**：变动小则残差小；临近到期或大动时迅速变大。
- **用途**：定位风险表缺哪一项；残差要解释，不能藏。

### 5.4 代码：归因



```python

dF, dvol, days = 1.0, 0.01, 1

exact = 0.0
for name, kind, strike, lots in positions:
    exact += (black76(F + dF, strike, T, vol + dvol, r, kind)
              - black76(F, strike, T, vol, r, kind)) * lots * mult

approx = (risk["现金Delta"].sum() * dF
          + 0.5 * risk["现金Gamma"].sum() * dF ** 2
          + risk["现金Vega"].sum() * dvol
          + risk["现金Theta"].sum() * days / 365)

print("精确重估", round(exact, 3))
print("希腊字母近似", round(approx, 3))
print("残差", round(exact - approx, 3))
```


**读数与应用**：残差应该不大；如果标的/波动率变动很大，残差会上升，需要加 Vanna/Volga 或直接重估。

**常见错误**：

1. 用“每 1 个点”的 Vega 乘 1.0 波动率。
2. Theta 用年值乘 1 天，忘记除以 365。
3. 只做一阶归因，标的大动时误差大。

**停下来想一想**：残差为负，说明什么？

<details><summary>对照你的思路</summary>

说明精确重估的收益低于希腊字母近似；可能是高阶项（Vanna/Volga/三阶）为负、或变动方向与交叉项相反。残差本身不是“错”，而是“近似没覆盖的部分”；要解释它来自哪一项。

</details>

## 6. Delta 对冲与残余风险

### 6.1 对冲手数

期货一手 Delta = 乘数（每点）。组合现金 Delta 为 $D$，对冲手数：

$$\text{lots}_{\text{hedge}}=-\frac{D}{\text{mult}}.$$

### 6.2 残余

Delta 对冲后：

- Gamma 不为零 → 标的大动时方向暴露回来；
- Vega 不为零 → 波动率变动仍亏/赚；
- Theta 仍在 → 时间流逝每天影响损益。

### 6.3 再平衡

Gamma 大时，Delta 变化快，要更频繁调仓；但调仓有成本。多久调一次，是“残余风险 vs 成本”的权衡。

### 6.4 代码：对冲手数与残余



```python

net_delta = risk["现金Delta"].sum()
hedge_lots = -net_delta / mult

print("组合现金 Delta", round(net_delta, 3))
print("对冲期货手数", round(hedge_lots, 3))
print("对冲后残余 Gamma", round(risk["现金Gamma"].sum(), 3))
print("对冲后残余 Vega", round(risk["现金Vega"].sum(), 3))
```


**读数与应用**：组合 Delta 为正/负，用期货反向对冲；Gamma/Vega 不会因为 Delta 对冲而消失。

**常见错误**：

1. 用“每单位 Delta”而不是“现金 Delta”算手数。
2. 忘记期货乘数。
3. 以为 Delta 中性就“没有风险”。

**停下来想一想**：Delta 中性组合，标的大涨或大跌，为什么都可能亏？

<details><summary>对照你的思路</summary>

Delta 中性只消掉一阶方向；Gamma 是二阶项。若 Gamma 为负（卖出期权），标的大涨或大跌都会亏；若 Gamma 为正，大动会赚。方向不确定时，Gamma 决定“波动越大越赚/越亏”。

</details>

## 本章小结

### 一张表

| 希腊字母 | 定义 | 单位 | 长仓看涨 |
|---|---|---|---|
| Delta | $\partial V/\partial F$ | 元/点 | 正 |
| Gamma | $\partial^2V/\partial F^2$ | Delta/点 | 正 |
| Vega | $\partial V/\partial\sigma$ | 元/波动率 | 正 |
| Theta | $\partial V/\partial t$ | 元/时间 | 负 |
| Vanna | $\partial^2V/\partial F\partial\sigma$ | Delta/波动率 | 正（平值） |
| Volga | $\partial^2V/\partial\sigma^2$ | Vega/波动率 | 正（平值） |

### 八句话

1. 希腊字母是偏导数，不是“风险名称”。
2. 单位比数值更重要。
3. 现金希腊字母 = 每单位 × 手数 × 乘数。
4. Delta 中性只消一阶方向。
5. Gamma 决定“大动时方向暴露回来”。
6. Vega 是波动率风险；不同期限要分桶。
7. Vanna/Volga 是二阶交叉项，大动时不能忽略。
8. 归因残差要解释，不要藏起来。

**1.** 买入看涨，Delta、Gamma、Vega、Theta 的符号是什么？

**2.** Vega 27.8（每 1.0 波动率），波动率涨 0.01，期权约变多少？

**3.** Theta −6.8/年，每过一天约变多少？

**4.** Vanna 的定义是什么？单位是什么？

**5.** 组合现金 Delta 55，期货乘数 10。对冲需要多少手？

**6.** 精确重估 10 元、希腊字母近似 9.2 元。残差多少？

**7.** 为什么 Delta 中性组合仍可能亏？

**8.** 为什么不同到期的 Vega 不能直接相加？

<details><summary>参考答案</summary>

**1.** Delta 正、Gamma 正、Vega 正、Theta 负。

**2.** $27.8\times0.01=0.278$ 元。

**3.** $-6.8/365\approx-0.0186$ 元/天。

**4.** $\partial^2V/\partial F\partial\sigma$；单位是“Delta/波动率单位”。

**5.** $-55/10=-5.5$ 手（方向相反，实际取 5 或 6 手）。

**6.** $10-9.2=0.8$ 元。

</details>
