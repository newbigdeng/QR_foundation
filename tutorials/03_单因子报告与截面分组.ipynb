## 1. 截面、因子值、预处理

### 1.1 截面

同一时刻、不同标的放在一起，叫截面。因子研究的第一步，是把“每个标的的因子值”和“未来收益”按同一天对齐：

| 日期 | 标的 | 因子值 | 未来收益 |
|---|---|---|---|
| 2026-11-02 | A | 1.2 | 0.8% |
| 2026-11-02 | B | 0.3 | −0.2% |
| 2026-11-02 | C | −0.5 | −1.1% |

同一行必须属于同一个截面、同一个预测起点。

### 1.2 预处理

- **缺失**：算不出来就留空，不要填 0；
- **异常值**：缩尾/截断，避免一个极端值主导；
- **标准化**：z-score 或排名，让不同因子可比；
- **中性化**：去掉市场/行业/规模暴露（进阶）。

### 1.3 排名 vs 标准化

- z-score 保留“相对幅度”，但对异常值敏感；
- 排名只保留“顺序”，对异常值稳健，但丢失幅度。

Rank IC 用排名，就是因为“排序”比“幅度”更稳健。

### 1.4 代码：异常值的影响



```python

import numpy as np
import pandas as pd

x = np.array([1.0, 2.0, 3.0, 100.0])
z = (x - x.mean()) / x.std(ddof=1)
r = pd.Series(x).rank(pct=True)

print("z-score", z.round(3))
print("百分位排名", r.round(3).to_list())
```


**读数与应用**：100 这个异常值把 z-score 拉到 1.5，其他三个都变成负数；排名仍然是 0.25/0.5/0.75/1.0，顺序清楚。因子研究里，先看分布，再决定缩尾还是排名。

**常见错误**：

1. 缺失填 0，把“没数据”误当成“中性”。
2. 不缩尾，一个极端值改变整列排序。
3. 用全样本均值/标准差标准化，再回测早期；用了未来信息。

**停下来想一想**：什么时候该用 z-score，什么时候该用排名？

<details><summary>对照你的思路</summary>

若因子的“幅度”有经济含义（比如估值偏离程度），用 z-score；若只关心排序、且异常值多，用排名。很多因子报告两者都看：Rank IC 看排序，分组收益看幅度。无论哪种，都要在训练段估计参数，再应用到测试段。

</details>

## 2. Rank IC：排序对不对

### 2.1 定义

同一天，把因子值和未来收益都换成排名，算相关系数：

$$IC_t=\operatorname{corr}(\operatorname{rank}(f_t),\operatorname{rank}(r_{t+1})).$$

**公式拆解**
- **符号**：$f_t$ 第 $t$ 日因子值，$r_{t+1}$ 下一期收益；$\operatorname{rank}$ 转名次，$N$ 只标的。
- **结构**：两列名次的 Pearson 相关（即 Spearman）；分母是两个名次标准差之积。
- **代入**：6 只，$d=(-1,1,-1,1,-1,1)$、$\sum d^2=6$ → $IC=1-\frac{36}{210}=0.829$。
- **边界**：名次全相同→分母 0→NaN，不能填 0；$N$ 太小单日 IC 无意义。
- **用途**：看排序能力；不是收益率，不含“差多远”的信息。

这就是 Rank IC（Spearman 相关）。取值 $[-1,1]$：

- 正：因子高，未来收益也高；
- 负：因子高，未来收益低（方向反了）；
- 0：没有排序能力。

### 2.2 手算

四只标的，因子排名 (1,2,3,4)，未来收益排名 (2,1,3,4)。

两列均值都是 2.5，去均值后：

| 因子 | 收益 | 去均值 f | 去均值 r | 乘积 |
|---:|---:|---:|---:|---:|
| 1 | 2 | −1.5 | −0.5 | 0.75 |
| 2 | 1 | −0.5 | −1.5 | 0.75 |
| 3 | 3 | 0.5 | 0.5 | 0.25 |
| 4 | 4 | 1.5 | 1.5 | 2.25 |

乘积和 4，平方和都是 5，所以

$$IC=\frac{4}{\sqrt{5\times5}}=0.8.$$

### 2.3 注意

- 同一天分数全相同：无法排序，IC 无定义，不能填 0；
- 并列：用平均排名；
- 样本太少（比如 4 只）：IC 波动很大，不能只看一天。

### 2.4 代码：Rank IC



```python

import pandas as pd

f = pd.Series([1, 2, 3, 4])
y = pd.Series([-1, -3, 0, 2])

print(pd.DataFrame({"因子": f, "未来收益": y, "收益排名": y.rank()}))
print("Rank IC", round(f.rank().corr(y.rank()), 3))
```


**读数与应用**：Rank IC 0.8，说明排序基本一致。注意：把整列收益减去 5，排名不变，IC 仍是 0.8——IC 只看排序，不看绝对收益。

**常见错误**：

1. 用 Pearson 相关算“排序能力”，异常值会主导。
2. 当天分数全相同还硬算 IC。
3. 只看一天 IC，不看时序稳定性。

**停下来想一想**：IC=0.8，是否意味着“买高分一定能赚”？

<details><summary>对照你的思路</summary>

不一定。IC 只说明“排序相关”，不说明幅度、成本、可交易性。若高分组只比低分组多 0.1%，扣掉双边成本可能净亏。IC 是“有没有排序能力”，不是“能不能赚钱”。

</details>

## 3. 分组收益：高分组真的更好吗

### 3.1 分组

按当天因子值排序，分成 $k$ 组（常用 3/5/10 组），算每组的未来收益。看：

- 是否单调：低→高组收益递增；
- 高减低：多空组合的毛收益；
- 每组样本数：别让某组只有一两个标的。

### 3.2 权重

- 等权：每组内每只标的权重相同；
- 市值/波动加权：更接近可投资组合；
- 多空：高组 +1、低组 −1，总敞口要写清。

### 3.3 一个陷阱

若高分组集中在某个行业/品种，分组收益可能只是“行业整体涨”，不是因子能力。要做行业中性/同期比较。

### 3.4 代码：三组



```python

import pandas as pd

df = pd.DataFrame({
    "score": [1, 2, 3, 4, 5, 6],
    "ret": [-1.0, -3.0, 0.0, 2.0, 1.0, 4.0],
})
df["group"] = pd.qcut(df["score"], 3, labels=["low", "mid", "high"])
print(df)
print(df.groupby("group", observed=True)["ret"].mean())
```


**读数与应用**：低组 −2，中组 1，高组 2.5；高减低 4.5。方向对、单调。但只有 6 个样本，不能据此下结论。

**常见错误**：

1. 分组后不报每组样本数。
2. 高减低用“总资金”还是“每边一半”说不清。
3. 不做行业/品种中性，把行业效应当因子。

**停下来想一想**：高减低 4.5%，是否等于“策略赚 4.5%”？

<details><summary>对照你的思路</summary>

不一定。高减低是“高组收益 − 低组收益”，可能是多空各一半资金，也可能各用 1 单位名义；还要扣双边成本、考虑换手、可成交性。报告要写清权重和成本口径。

</details>

## 4. 换手与成本：扣费后还剩多少

### 4.1 换手

相邻两期权重变化：

$$\text{turnover}_t=\sum_i|w_{t,i}-w_{t-1,i}|.$$

**公式拆解**
- **符号**：$w_{t,i}$ 第 $t$ 期标的 $i$ 权重（占净值），$|\cdot|$ 是绝对值；双边口径。
- **结构**：逐只算仓位变化绝对值再求和=路径长度；权重和不变时单边=双边/2。
- **代入**：$(0.4,0.3,0.2,0.1)\to(0.2,0.5,0.3,0)$ → $0.2+0.2+0.1+0.1=0.6$。
- **边界**：全清仓再买回=2；价格漂移造成的权重变化不是交易。
- **用途**：成本与容量估算；不等于交易次数，双边/单边不能混报。

若多空各占一半，单边换手 1.0 表示“整个组合换了一遍”。报告要写清单边/双边。

### 4.2 成本

$$\text{net}=w^Tr-\text{fee}\times\text{turnover}.$$

**公式拆解**
- **符号**：$w$ 权重，$r$ 下期收益（小数），$\text{fee}$ 单位换手成本，$\text{turnover}$ 换手。
- **结构**：净收益=毛利 $- \text{fee}\times$换手；两条腿同为收益率，可直接相减。
- **代入**：毛利 95bp、换手 0.6、单边 5bp → 成本 3bp，净 92bp，保本换手 19。
- **边界**：$\text{fee}$ 与换手必须同口径（否则差 2 倍）；多期要复利。
- **用途**：扣费后收益与容量；毛 IC 不能直接当净利。

费用包括手续费、价差、滑点。高频/高换手因子，成本会吃掉大部分毛利。

### 4.3 手算

旧权重 $(0.5,0.5,0,0)$，新权重 $(0,0.5,0.5,0)$：

$$\text{turnover}=0.5+0+0.5+0=1.0.$$

若单边成本 5bp，成本 $=1.0\times0.0005=0.0005=5$bp。

### 4.4 代码：换手与成本



```python

import numpy as np

w_old = np.array([0.5, 0.5, 0.0, 0.0])
w_new = np.array([0.0, 0.5, 0.5, 0.0])

turnover = np.abs(w_new - w_old).sum()
cost = turnover * 0.0005
print("换手", turnover, "成本", cost)
```


**读数与应用**：换手 1.0、成本 5bp。若因子日频调仓、单边成本 5bp，一年 250 天，成本可能超过 100%，必须看净收益。

**常见错误**：

1. 只算手续费，不算价差/滑点。
2. 单边/双边混用。
3. 用“持仓不变”假设算换手，忽略权重漂移。

**停下来想一想**：为什么高 IC 因子也可能被换手拖垮？

<details><summary>对照你的思路</summary>

IC 高只说明排序好；如果排序每天大幅变化，组合要频繁调仓，成本累积。一个 IC 0.05、日换手 100% 的因子，扣费后可能不如 IC 0.03、月换手 20% 的因子。**净收益 = 毛利 − 成本**，两者要一起看。

</details>

## 5. 逐日 IC 与 ICIR：稳不稳定

### 5.1 从一天到一段时间

每天算一个 IC，得到序列 $IC_1,\dots,IC_T$。看：

$$\bar{IC}=\frac1T\sum_t IC_t,\qquad s_{IC}=\sqrt{\frac{1}{T-1}\sum_t(IC_t-\bar{IC})^2}.$$

t 值：

$$t=\frac{\bar{IC}}{s_{IC}/\sqrt T}.$$

ICIR：

$$\text{ICIR}=\frac{\bar{IC}}{s_{IC}}.$$

**公式拆解**
- **符号**：$IC_t$ 逐日 Rank IC，$\bar{IC}$ 均值，$s_{IC}$ 标准差，$T$ 天数。
- **结构**：单位波动的预测力；与 $t$ 只差样本杠杆：$t=\text{ICIR}\sqrt T$。
- **代入**：$\bar{IC}=0.04$、$s_{IC}=0.10$ → ICIR $=0.40$；$T=120$ 时 $t=4.38$。
- **边界**：$s_{IC}=0$ 无定义；IC 自相关时 $t=\text{ICIR}\sqrt T$ 高估显著性。
- **用途**：因子排序、ICIR 加权；不是 Sharpe。

ICIR 是“单位波动的预测力”；年化要乘 $\sqrt{252}$（日频）或 $\sqrt{12}$（月频）。

### 5.2 自相关

IC 序列常有自相关；普通 t 值会高估显著性。实务中用 Newey–West 调整，或块状 bootstrap。

### 5.3 手算

$\bar{IC}=0.03$，$s=0.08$，$T=60$：

$$t=\frac{0.03}{0.08/\sqrt{60}}\approx2.90,\qquad \text{ICIR}=0.375.$$

单看 t 值“显著”，但只有 60 天，且 IC 可能自相关；不能直接下结论。

### 5.4 代码：IC 的 t 值与 ICIR



```python

import numpy as np

rng = np.random.default_rng(301)
raw = rng.normal(0, 1, size=60)
ic = 0.03 + (raw - raw.mean()) * 0.08 / raw.std(ddof=1)

t_stat = ic.mean() / (ic.std(ddof=1) / np.sqrt(len(ic)))
print("IC 均值", round(ic.mean(), 4))
print("IC 标准差", round(ic.std(ddof=1), 4))
print("t 值", round(t_stat, 3), "ICIR", round(ic.mean() / ic.std(ddof=1), 3))
```


**读数与应用**：t 值约 2.9、ICIR 0.375。t 值告诉你“均值是否可能为 0”，ICIR 告诉你“单位波动的预测力”。两个都要看，不能只看一个。

**常见错误**：

1. 用日 IC 的 t 值直接年化显著性，忽略自相关。
2. 把 ICIR 当夏普。ICIR 是 IC 的均值/标准差，不是组合收益。
3. 只看均值，不看 IC 的时序图；少数几天可能撑起全部。

**停下来想一想**：IC 均值 0.03、t 值 2.9，是否说明因子很好？

<details><summary>对照你的思路</summary>

不一定。IC 0.03 很小；t 值 2.9 只说明“不容易是 0”，不说明经济价值。还要看分组收益、换手、成本、样本外。一个统计显著但扣费后净亏的因子，不是好因子。

</details>

## 6. 标签尺度与重叠：别把 5 天标签当 5 个独立样本

### 6.1 标签期限

- 日频标签：预测下一天；
- 5 日标签：预测未来 5 天；
- 20 日标签：预测未来 20 天。

标签期限要和持仓周期、换手成本匹配。用 5 日标签，却每天调仓，成本和标签不匹配。

### 6.2 重叠

若每天发信号、标签是未来 5 天，相邻样本共享 4 天行情：

| 信号日 | 标签窗口 |
|---|---|
| 周一 | 周二–下周一 |
| 周二 | 周三–下周二 |

有效独立样本大约是 $T/5$，不是 $T$。t 值会被高估。

### 6.3 处理

- 按标签期限抽样（每 5 天取一个）；
- 用块状 bootstrap；
- 用 purged/embargo 切分；
- 报告“独立交易日数”，不只报样本行数。

### 6.4 代码：重叠窗口



```python

T, horizon = 10, 5
for t in range(T):
    window = list(range(t + 1, min(t + horizon, T - 1) + 1))
    print("信号日", t, "标签窗口", window)
```


**读数与应用**：相邻信号的标签窗口大量重叠；样本行数 10，独立窗口大约只有 2 个。报告里要写“有效独立样本”，不能只写行数。

**常见错误**：

1. 用行数当样本量，t 值虚高。
2. 日频调仓、5 日标签，成本和标签不匹配。
3. 随机 K 折，重叠标签跨训练/测试。

**停下来想一想**：如果标签是未来 5 天，但策略每月调仓，应该怎么对齐？

<details><summary>对照你的思路</summary>

按月生成信号、按月计算标签，或把日频信号聚合成月频组合；不要让“每天的信号”与“5 天标签”混在一起。标签期限、调仓频率、成本口径要一致，否则 t 值和净收益都不可信。

</details>

## 7. 一份最小因子报告

### 7.1 六块

| 块 | 内容 | 要回答 |
|---|---|---|
| 覆盖率 | 每天多少标的能算因子、缺失分布 | 样本够不够 |
| IC | 均值、标准差、t、ICIR、时序图 | 排序稳不稳 |
| 分组 | 3/5 组收益、单调性、高减低 | 高分组是否更好 |
| 换手 | 单边/双边、日均换手 | 成本高不高 |
| 成本 | 手续费、价差、滑点、净收益 | 扣费后剩多少 |
| 样本外 | 训练/测试切分、独立交易日 | 是否可迁移 |

### 7.2 代码：合成面板的最小报告



```python

import numpy as np
import pandas as pd

rng = np.random.default_rng(302)
T, N = 80, 20
score = pd.DataFrame(rng.normal(size=(T, N)))
future = 0.002 * score + pd.DataFrame(rng.normal(0, 0.01, size=(T, N)))

ic = score.rank(axis=1).corrwith(future.rank(axis=1), axis=1)
groups = (((score.rank(axis=1) - 1) * 3) // N).astype(int)
group_ret = pd.DataFrame({g: future.where(groups == g).mean(axis=1) for g in range(3)})

w = ((groups == 2).astype(float) - (groups == 0).astype(float)) / (2 * (N // 3))
turn = w.diff().abs().sum(axis=1)
turn.iloc[0] = w.iloc[0].abs().sum()
net = (w * future).sum(axis=1) - 0.0005 * turn

print("IC 均值/标准差/t", round(ic.mean(), 4), round(ic.std(ddof=1), 4), round(ic.mean() / (ic.std(ddof=1) / np.sqrt(T)), 3))
print("分组平均收益", group_ret.mean().round(5).to_dict())
print("日均换手", round(turn.mean(), 3), "日均净收益", round(net.mean(), 5))
```


**读数与应用**：合成面板的 IC 约 0.17（合成数据偏高）、分组单调、净收益为正；这是“教具”，不是真实市场。真实报告要加样本外、行业中性、成本敏感性。

**常见错误**：

1. 只报 IC，不报分组/换手/成本。
2. 用全样本选因子，再报同一段 IC。
3. 不报覆盖率；缺失可能集中在某些品种/时段。

**停下来想一想**：一份报告里，哪个数最能证伪“因子有效”？

<details><summary>对照你的思路</summary>

样本外净收益和“被拒组影子账”。样本内 IC 可以有很多解释；样本外扣费后仍为正、且被拒组确实更差，才比较难反驳。单一 IC 数字最容易被多重检验/选择偏差污染。

</details>

## 8. 常见错误与检查清单

### 8.1 五个常见错误

1. **混截面**：把不同日期/不同品种混在一起算 IC。
2. **未来函数**：用当天收盘算因子，又用当天收盘成交。
3. **重叠标签**：5 日标签当独立样本。
4. **忽略成本**：只看毛利/IC。
5. **多重检验**：试了 50 个因子，只报最好的。

### 8.2 检查清单

- [ ] 因子只用 $t$ 及之前的数据；
- [ ] 标签窗口明确，重叠样本标记；
- [ ] 分日期算 IC，不混截面；
- [ ] 分组报样本数、单调性、高减低；
- [ ] 换手/成本口径写清；
- [ ] 训练/测试按时间切；
- [ ] 报覆盖率、独立交易日；
- [ ] 结论能被反例推翻。

## 本章小结

### 一张表

| 指标 | 公式 | 看什么 |
|---|---|---|
| Rank IC | $\operatorname{corr}(\operatorname{rank}(f),\operatorname{rank}(r))$ | 排序能力 |
| 分组收益 | 组内平均未来收益 | 单调性/高减低 |
| 换手 | $\sum|w_t-w_{t-1}|$ | 成本 |
| 净收益 | $w^Tr-\text{fee}\times\text{turnover}$ | 扣费后 |
| IC t 值 | $\bar{IC}/(s/\sqrt T)$ | 显著性 |
| ICIR | $\bar{IC}/s$ | 单位波动预测力 |

### 八句话

1. 同一行必须同一截面、同一预测起点。
2. Rank IC 只看排序，不看幅度。
3. IC 无定义时不能填 0。
4. 分组要看单调性、样本数、行业暴露。
5. 高 IC 不等于净赚钱；成本和换手会吃掉毛利。
6. IC 的 t 值要经得起自相关/重叠样本。
7. 5 日标签的有效样本约是行数的 1/5。
8. 报告要能被人按同样规则复查。
