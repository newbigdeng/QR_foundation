## 1. 五个状态

| 状态 | 含义 | 进入条件 | 退出 |
|---|---|---|---|
| 开发 | 研究/回测 | 有假设/数据 | 通过验证 |
| 冻结观察 | 参数锁定，等前向 | 验证通过 | 完整交易日达标 |
| 可用 | 可上线/小仓 | 前向达标 | 触发暂停 |
| 暂停 | 停止开仓/减仓 | 异常 | 修复/退役 |
| 退役 | 不再使用 | 失效确认 | — |

**状态不是“进度条”，是责任边界。** 每个状态要有进入/退出条件、负责人、记录。

**常见错误**：

1. 跳过冻结观察直接上线。
2. 亏损就改参数，不区分输入/信号/执行。
3. 没有退役条件，因子永远“在观察”。

**停下来想一想**：为什么“开发→可用”不能一步到位？

<details><summary>对照你的思路</summary>

开发阶段用的是历史数据，已经参与研究/调参；直接上线等于把样本内当样本外。冻结观察用“冻结后真实发生”的样本，才验证可迁移性。

</details>

## 2. 失效定位：输入、信号、执行

### 2.1 三层

- **输入**：数据缺失/修订/合约映射错误 → 可见样本变了；
- **信号**：同样的输入，因子与未来收益的关系变弱；
- **执行**：成本/滑点/拒单变差 → 毛利没变，净利变差。

### 2.2 拆净利

$$\text{净利}=\text{毛利}-\text{成本}.$$

**公式拆解**
- **符号**：毛利=价差×乘数×手数（元，未扣费）；成本=手续费+价差+滑点（元，双边计）。
- **结构**：$\text{净利}=\text{毛利}-\text{成本}$；写成变化量 $\Delta$净利$=\Delta$毛利$-\Delta$成本 才能定位来源。
- **代入**：毛利 8、成本 3 → 净利 5；成本升到 10 → 净利 $-2$，变化 $-7$。
- **边界**：成本>毛利则净利为负；成本可为负（返佣），净利可高于毛利。
- **用途**：失效定位第一刀（执行 vs 信号）；不是现金流，也不证明因子有效。

若毛利不变、成本上升，是执行问题；若毛利下降、成本不变，是信号问题；若覆盖率/样本变了，先查输入。

### 2.3 代码：毛利/成本拆解



```python

before = {"gross": 8.0, "cost": 3.0}
after = {"gross": 8.0, "cost": 10.0}

print("净利变化", (after["gross"] - after["cost"]) - (before["gross"] - before["cost"]))
print("毛利变化", after["gross"] - before["gross"])
print("成本变化", after["cost"] - before["cost"])
```


**读数与应用**：毛利不变、成本从 3 升到 10，净利从 +5 变 −2；这是执行问题，不是信号失效。

**常见错误**：

1. 只看净利，不拆毛利/成本。
2. 把成本上升当“因子失效”。
3. 不查覆盖率/样本变化。

**停下来想一想**：毛利不变、成本上升，应该先改因子还是先查执行？

<details><summary>对照你的思路</summary>

先查执行：价差、滑点、手续费、拒单、成交延迟。因子没变，改因子会掩盖执行问题。

</details>

## 3. 技术性暂停 vs 统计性暂停

### 3.1 两种暂停

- **技术性暂停**：数据/系统/执行出问题，暂停止损，修好再开；
- **统计性暂停**：样本显示原假设不再成立，需要新研究/更多样本。

### 3.2 规则

事前写：

- 连续多少笔/多少日净利为负；
- 回撤超过多少；
- 数据缺失超过多少；
- 触发后是“暂停”还是“退役”。

### 3.3 代码：连续亏损计数



```python

streak = 0
for value in [-2.0, None, -1.0, -3.0]:
    if value is None:
        pass
    elif value < 0:
        streak += 1
    else:
        streak = 0
    print(value, "连续亏损", streak, "触发复核" if streak >= 2 else "继续记录")
```


**读数与应用**：缺失值不打断连续亏损计数（按事先规则），连续 2 次触发复核。规则要在看结果前写死。

**常见错误**：

1. 看到亏损临时改规则。
2. 缺失值当 0。
3. 只暂停不记录原因。

**停下来想一想**：为什么“暂停规则”要事前写死？

<details><summary>对照你的思路</summary>

事后改规则会变成“看结果再决定”，产生选择偏差；事前写死才能保证规则的一致性/可复查。

</details>

## 4. 拒绝样本与影子账

### 4.1 影子账

被过滤器拒绝的信号，按同一套入场/退出/成本规则记“影子盈亏”；和实际账户分开。

### 4.2 评价

- 通过组 vs 拒绝组；
- 同日配对；
- 样本量/独立交易日。

**公式拆解**
- **符号**：$\pi_k$ 第 $k$ 个信号净利（元，同规则记账），$\bar\pi_1$、$\bar\pi_0$ 为通过/拒绝组均值。
- **结构**：区分力 $\Delta=\bar\pi_1-\bar\pi_0$；同日配对差再求均值与 $t$ 值。
- **代入**：通过 $(3,-1)$ 均值 1、拒绝 $(-4,2)$ 均值 $-1$ → $\Delta=2$ 元/笔。
- **边界**：两组样本不等要换成每笔；拒绝组没记账则无法算。
- **用途**：评价过滤器增量；不是绝对盈利，小样本也不算显著。

### 4.3 代码：通过/拒绝



```python

import numpy as np

pnl = np.array([-4.0, 3.0, -1.0, 2.0])
keep = np.array([False, True, True, False])

print("通过合计", pnl[keep].sum(), "拒绝合计", pnl[~keep].sum())
print("通过均值", pnl[keep].mean(), "拒绝均值", pnl[~keep].mean())
```


**读数与应用**：通过组均值 1，拒绝组均值 −1；过滤器有区分力。但样本只有 4 笔，不能下结论。

**常见错误**：

1. 只看通过组。
2. 不记录拒绝原因。
3. 样本太小就宣称有效。

**停下来想一想**：为什么“没有交易”不等于“没有亏损”？

<details><summary>对照你的思路</summary>

被拒绝的信号可能本来会赚/会亏；不记影子账，就不知道过滤器是“避开了亏损”还是“错过了盈利”。没有交易只是“没进场”，不是“没结果”。

</details>

## 6. 因子寿命与四道门

统计套利因子的有效窗口常被估计在 6–18 个月，A 股因子生命周期常在 3–5 年；衰减是常态，不是例外。要区分：

- **周期性衰减**：风格/宏观暂时切换，可能恢复；
- **结构性衰减**：逻辑被套利/规则改变，难以恢复。

退役前过四道门：逻辑门、样本门、交易门、风险门；再用滚动 IC/CUSUM 监控。不要因为一个月亏损就退役，也不要因为“曾经有效”就一直留着。

## 本章小结

### 一张表

| 概念 | 做法 | 用途 |
|---|---|---|
| 五状态 | 开发/冻结/可用/暂停/退役 | 责任边界 |
| 三层失效 | 输入/信号/执行 | 定位 |
| 技术性暂停 | 数据/系统 | 修复 |
| 统计性暂停 | 样本/假设 | 新研究 |
| 影子账 | 拒绝信号同规则记账 | 评价过滤器 |

### 八句话

1. 状态要有进入/退出条件。
2. 不要跳过冻结观察。
3. 净利 = 毛利 − 成本，先拆再判断。
4. 毛利不变、成本升，先查执行。
5. 技术性暂停修系统，统计性暂停做研究。
6. 暂停规则事前写死。
7. 拒绝信号要记影子账。
8. 小样本不能下强结论。
