## 1. 资产类别：先看收益、风险、相关性

### 1.1 四类资产

| 资产 | 收益来源 | 主要风险 | 流动性 |
|---|---|---|---|
| 现金 | 利率 | 通胀/机会成本 | 高 |
| 债券 | 票息/利率下行 | 利率/信用 | 中高 |
| 股票 | 盈利增长/估值 | 权益/回撤 | 高 |
| 商品 | 供需/展期 | 波动/展期/交割 | 中 |

### 1.2 相关性

股票和债券常负相关/低相关，是 60/40 的基础；商品和股票相关性不稳定；危机时相关性可能同时上升。

### 1.3 代码：资产统计



```python

import numpy as np

names = ["现金", "债券", "股票", "商品"]
mu = np.array([0.02, 0.03, 0.07, 0.05])
vol = np.array([0.01, 0.05, 0.16, 0.18])
corr = np.array([
    [1.0, 0.1, 0.0, 0.0],
    [0.1, 1.0, -0.1, 0.0],
    [0.0, -0.1, 1.0, 0.3],
    [0.0, 0.0, 0.3, 1.0],
])
cov = np.outer(vol, vol) * corr
for i, name in enumerate(names):
    print(name, "收益", f"{mu[i]:.1%}", "波动", f"{vol[i]:.1%}")
```


**读数**：股票收益最高、波动最大；债券波动低、与股票负相关；商品波动高、与股票正相关。**配置不是选收益最高，而是选“组合起来”的效率。**

**常见错误**：

1. 用单资产历史收益排序。
2. 忽略相关性/危机相关性。
3. 把商品当“股票替代”。

## 2. 战略配置：IPS 与风险预算

### 2.1 最小 IPS

- 目标：保值/增值/现金流；
- 期限：1 年/5 年/10 年；
- 风险容忍：最大回撤/波动上限；
- 基准：60/40、CPI+、同行；
- 约束：流动性、杠杆、集中度、ESG。

### 2.2 风险预算

不是“资金预算”，而是“风险预算”：给每类资产分配风险贡献，而不是权重。

### 2.3 代码：风险预算



```python

w_eq = np.array([0.0, 0.4, 0.6, 0.0])
port_vol = np.sqrt(w_eq @ cov @ w_eq)
mrc = (cov @ w_eq) / port_vol
rc = w_eq * mrc / port_vol
print("60/40 波动", f"{port_vol:.2%}", "风险贡献", np.round(rc, 3))
```


**读数**：60/40 的波动约 9.6%，风险贡献几乎全来自股票（0.98），债券只贡献 0.02。**资金 40% 的债券只贡献 2% 的风险。**

**常见错误**：

1. 把权重当风险预算。
2. 不做风险贡献分解。
3. 忽略再平衡/相关性变化。

## 3. 配置方法：60/40、等权、均值方差、风险平价

### 3.1 四种方法

- 60/40：简单、基准；
- 等权：分散，但风险集中；
- 均值方差：理论最优，但对 $\mu$ 敏感；
- 风险平价：按风险贡献相等，常与波动率倒数近似。

### 3.2 代码：权重与波动



```python

def port_stats(w):
    return float(w @ mu), float(np.sqrt(w @ cov @ w))

methods = {
    "60/40": np.array([0.0, 0.4, 0.6, 0.0]),
    "等权": np.array([0.25, 0.25, 0.25, 0.25]),
    "风险平价(逆波动)": (1 / vol) / (1 / vol).sum(),
}
for name, w in methods.items():
    ret, sd = port_stats(w)
    print(f"{name}: 权重 {np.round(w,3)}, 收益 {ret:.2%}, 波动 {sd:.2%}")
```


**读数**：风险平价把更多权重给低波动资产，组合波动通常低于 60/40；收益也更低。**没有免费午餐，只有风险/收益的交换。**

**常见错误**：

1. 只看收益排序选方法。
2. 风险平价不做杠杆/约束。
3. 均值方差用原始 $\mu$ 不加收缩。

## 4. 再平衡：日历 vs 阈值

### 4.1 两种规则

- 日历：每月/每季/每年再平衡；
- 阈值：权重偏离目标超过 5% 再平衡。

### 4.2 成本

再平衡有交易成本；越频繁，成本越高；越不频繁，漂移越大。

### 4.3 代码：模拟



```python

def simulate_rebalance(rule="annual", years=20, cost=0.001, seed=3301):
    rng = np.random.default_rng(seed)
    n = years * 12
    rets = rng.normal(mu / 12, vol / np.sqrt(12), size=(n, 4))
    w = np.array([0.0, 0.4, 0.6, 0.0])
    values = np.ones(4) * w
    total_cost = 0.0
    for t in range(n):
        values = values * (1 + rets[t])
        total = values.sum()
        w_now = values / total
        rebalance = False
        if rule == "annual" and (t + 1) % 12 == 0:
            rebalance = True
        if rule == "threshold" and np.abs(w_now - w).max() > 0.05:
            rebalance = True
        if rebalance:
            trade = np.abs(w * total - values).sum()
            total_cost += trade * cost
            values = w * (total - trade * cost)
    return values.sum(), total_cost

for rule in ["annual", "threshold"]:
    terminal, cost = simulate_rebalance(rule)
    print(rule, "终值", round(terminal, 4), "累计成本", round(cost, 4))
```


**读数**：本例阈值再平衡成本略高（0.0023 vs 0.0019）、终值略高（1.886 vs 1.857）；没有绝对最优，取决于阈值/波动/成本。**日历再平衡更简单、更可审计。**

**常见错误**：

1. 再平衡忽略成本。
2. 阈值用“权重差”而不是“风险差”。
3. 危机时机械再平衡，忽略流动性。

## 5. 业绩评价：Sharpe、IR、最大回撤

### 5.1 指标

$$\text{Sharpe}=\frac{E[r-r_f]}{\sigma},\qquad \text{IR}=\frac{E[r-b]}{\sigma(r-b)}.$$

**公式拆解**
- **符号**：$r$ 是组合在统计区间内的平均收益率（年化小数），$r_f$ 是无风险利率，$b$ 是基准收益率，$\sigma$ 是组合收益标准差，$\sigma(r-b)$ 是相对基准超额收益的跟踪误差，五个量必须同为年化或同为月频口径。
- **结构**：Sharpe 的分子是相对无风险利率的超额收益、分母是总波动，IR 的分子是相对基准的超额收益、分母是跟踪误差，基准取无风险利率时 IR 恰好退化为 Sharpe。
- **代入**：$r=12\%$、$r_f=2\%$、$\sigma=16\%$ 得 Sharpe $(12-2)/16=0.625$，若再取 $b=8\%$、跟踪误差 6% 则 IR 为 $(12-8)/6\approx0.667$。
- **边界**：$\sigma\to0$ 时分母为 0 指标无定义，跟踪误差为 0 时 IR 也无定义，超额收益为负时得到负指标而不是 0。
- **用途**：用同一把尺子做策略筛选、同类比较与业绩沟通；它不是收益预测也不是 alpha 证据，挑过的策略还要做多重检验（DSR/PBO）校正。

最大回撤：净值从峰值到谷底的最大跌幅。

### 5.2 代码：指标



```python

def metrics(nav):
    ret = np.diff(nav) / nav[:-1]
    sharpe = ret.mean() / ret.std(ddof=1) * np.sqrt(12)
    peak = np.maximum.accumulate(nav)
    mdd = (nav / peak - 1).min()
    return sharpe, mdd

nav = np.cumprod(1 + np.random.default_rng(3302).normal(0.006, 0.03, 240))
sharpe, mdd = metrics(np.r_[1.0, nav])
print("年化 Sharpe", round(sharpe, 3), "最大回撤", f"{mdd:.2%}")
```


**读数**：本例 Sharpe 约 0.74、最大回撤约 −16.6%；高 Sharpe 不代表低回撤，两个都要报。

**常见错误**：

1. 只报收益/Sharpe，不报回撤。
2. 用月度/日度混算年化。
3. 不跟基准比。

## 6. 费用、容量与治理

### 6.1 费用拖累

管理费 1%/年，30 年吃掉约 26% 的终值（未计超额收益）；高换手还有交易成本。

### 6.2 容量

规模越大，冲击成本越高；容量曲线通常先升后降。

### 6.3 治理

- 投资委员会/授权；
- 基准/风险限额；
- 再平衡/偏离权限；

## 本章小结

### 一张表

| 概念 | 做法 | 用途 |
|---|---|---|
| IPS | 目标/期限/风险/基准 | 顶层约束 |
| 60/40 | 40 债 60 股 | 基准 |
| 风险平价 | 风险贡献相等 | 分散 |
| 再平衡 | 日历/阈值 | 控漂移 |
| Sharpe/IR | 超额/波动 | 评价 |
| 费用 | 年费×年限 | 长期拖累 |

### 八句话

1. 配置不是选收益最高。
2. 权重不等于风险贡献。
3. 60/40 的风险几乎全来自股票。
4. 风险平价用低波动资产降波动。
5. 再平衡是纪律，也有成本。
6. Sharpe 高不代表回撤小。
7. 费用/容量会吃掉长期收益。
8. IPS 是配置的“宪法”。
