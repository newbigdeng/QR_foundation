## 1. 导数：斜率、极限、Delta

### 1.1 定义

函数 $f(x)$ 在 $x$ 点的导数：

$$f'(x)=\lim_{h\to0}\frac{f(x+h)-f(x)}{h}.$$

几何上，它是曲线在这一点切线的斜率；交易上，它是“输入动一点，输出动多少”。

$$f'(x)\approx\frac{f(x+h)-f(x-h)}{2h}$$

叫中心差分，比单边差分更准。

### 1.2 期权的 Delta

设期权价值 $V(F)$，标的价格 $F$。Delta：

$$\Delta=\frac{\partial V}{\partial F}.$$

**公式拆解**
- **符号**：$V$ 期权价值、$F$ 标的价；$\Delta$ 无量纲（元/元）。
- **结构**：$\partial V/\partial F$ 是敏感度，也是“1 张期权 $\approx\Delta$ 张标的”的对冲比例。
- **代入**：$V(99)=4.55$、$V(100)=5.00$、$V(101)=5.55$ → $\Delta=\frac{1.00}{2}=0.50$。
- **边界**：$\Delta$ 随价格变（Gamma）；临近到期时平值 $\Delta$ 从 0.5 跳向 0/1。
- **用途**：对冲手数与敞口；不是到期概率，中性≠无风险。

它回答“标的涨 1 元，期权涨多少”。看涨 Delta 在 0 到 1 之间，看跌在 −1 到 0 之间。

### 1.3 手算

$f(x)=x^2$，在 $x=3$：

- 解析导数 $2x=6$；
- 中心差分，$h=0.001$：$[(3.001)^2-(2.999)^2]/(0.002)\approx6.000$。

步长太大，差商偏离；步长太小，浮点误差放大。通常 $h\approx10^{-4}$ 到 $10^{-6}$，要看函数尺度。

### 1.4 代码：差商



```python

import numpy as np

def f(x):
    return x ** 2

x, h = 3.0, 1e-4
print("前向差商", (f(x + h) - f(x)) / h)
print("中心差商", (f(x + h) - f(x - h)) / (2 * h))
print("解析导数", 2 * x)
```


**读数与应用**：中心差商应该非常接近 6。若差很多，先检查 $h$ 是否合适，再检查函数是否在这一点可导。

**常见错误**：

1. 用太大的 $h$。差商变成“割线斜率”，不是切线。
2. 用太小的 $h$。两个相近的数相减，浮点误差被放大。
3. 把“导数”当成“变化量”。导数是斜率，变化量还要乘上输入的变动。

**停下来想一想**：为什么期权 Delta 不是常数？

<details><summary>对照你的思路</summary>

因为期权价值对标的不是直线，而是曲线。标的越涨，看涨期权越可能变成实值，Delta 从 0 附近升到 1 附近；标的越跌，Delta 越小。Delta 随标的变化，这个变化率就是 Gamma。

</details>

### 1.5 三个必须会的求导规则

- 幂函数：$(x^n)'=nx^{n-1}$；
- 乘积：$(fg)'=f'g+fg'$；
- 链式：$f(g(x))'=f'(g(x))g'(x)$。

期权里最常用的是链式法则：$V(F(t))$ 对 $t$ 求导，要乘 $F'(t)$。例如 $V=\ln F$，$V_F=1/F$，$V_{FF}=-1/F^2$；这两个式子在第06章的伊藤引理里会再出现。

## 2. 二阶导数：曲率、Gamma

### 2.1 定义

二阶导数是一阶导数的导数：

$$f''(x)=\frac{d}{dx}f'(x).$$

它描述“斜率变化的速度”，也就是曲线的弯曲程度。

中心二阶差分：

$$f''(x)\approx\frac{f(x+h)-2f(x)+f(x-h)}{h^2}.$$

### 2.2 期权的 Gamma

$$\Gamma=\frac{\partial^2 V}{\partial F^2}.$$

**公式拆解**
- **符号**：$V$ 期权价值、$F$ 标的价；$\Gamma=V_{FF}=d\Delta/dF$，单位 1/元。
- **结构**：$\Delta$ 的变化率，泰勒二阶项 $\frac12\Gamma(dF)^2$；买方正、卖方负。
- **代入**：$5.55-2\times5.00+4.55=0.10$ → $\Gamma=0.10$，$1$ 元涨幅的二阶损益 $0.05$ 元。
- **边界**：临近到期平值 $\Gamma$ 爆炸；步长小于最小变动价位时全是舍入噪声。
- **用途**：定对冲频率、Gamma 交易；不能当风险大小，跳空时泰勒失效。

它回答“标的涨 1 元，Delta 变多少”。Gamma 大，说明 Delta 变得快，对冲要更频繁。

看涨、看跌的 Gamma 都是正的（买方）；卖方为负。平值附近 Gamma 最大，临近到期更尖。

### 2.3 手算

期权在 $F=99,100,101$ 时价值 $4.55,5.00,5.55$：

$$\Delta\approx\frac{5.55-4.55}{2}=0.50,$$

$$\Gamma\approx\frac{5.55-2\times5.00+4.55}{1^2}=0.10.$$

标的从 100 涨到 101，Delta 从约 0.5 升到约 0.6；这就是 Gamma 的效果。

### 2.4 代码：二阶差分



```python

import numpy as np

def g(x):
    return x ** 2

x, h = 3.0, 1e-3
print("二阶差分", (g(x + h) - 2 * g(x) + g(x - h)) / h ** 2)
print("解析二阶导", 2.0)
```


**读数与应用**：$x^2$ 的二阶导恒为 2；二阶差分应该接近 2。若差很多，通常是 $h$ 太大或函数尺度太大。

**常见错误**：

1. 把 Gamma 当成“风险的大小”。Gamma 是曲率，不是波动率；它衡量 Delta 的变化速度。
2. 忽略 Gamma 的符号。买方正 Gamma，卖方负 Gamma；负 Gamma 在标的大动时会亏。
3. 用一阶差分算二阶导。误差会很大，尤其是噪声数据。

**停下来想一想**：为什么平值期权的 Gamma 最大？

<details><summary>对照你的思路</summary>

平值时，期权最“犹豫”：标的小涨可能变实值，小跌可能变虚值，Delta 在 0.5 附近变化最快。深度实值时 Delta 接近 1，深度虚值时接近 0，两者都“变不动”，所以 Gamma 小。临近到期时，这种“犹豫”集中在很窄的价格区间，Gamma 更尖。

</details>

### 2.5 Gamma 与 Theta 的权衡

买方正 Gamma、负 Theta：标的动得越大越赚，但每天要付时间价值。卖方负 Gamma、正 Theta：横盘每天赚，但标的大动会亏。

所以“买期权还是卖期权”不是看方向，而是看：

- 你预期实际波动率比隐含波动率高还是低；
- 你能不能承受负 Gamma 的尾部风险；
- 你的对冲频率和成本。

Gamma 和 Theta 是一对：想要正 Gamma，通常要付 Theta；想要收 Theta，通常要承担负 Gamma。

## 3. 偏导数与时间：Theta

### 3.1 偏导数

期权价值依赖多个输入：$V(F,t,\sigma,r)$。固定其他输入，只动一个，得到偏导数：

$$\frac{\partial V}{\partial F},\quad\frac{\partial V}{\partial t},\quad\frac{\partial V}{\partial \sigma}.$$

全微分：

$$dV=\frac{\partial V}{\partial F}dF+\frac{\partial V}{\partial t}dt+\frac{\partial V}{\partial \sigma}d\sigma+\dots$$

这就是“每个输入动一点，输出怎么变”的线性叠加。

### 3.2 Theta

$$\Theta=\frac{\partial V}{\partial t}.$$

注意时间方向：$t$ 是日历时间，$\tau=T-t$ 是剩余期限。$\partial V/\partial t=-\partial V/\partial\tau$。很多教材的 Theta 是“每过一天，期权价值变多少”，符号容易混。

### 3.3 手算

设 $V=2\sqrt{\tau}$，$\tau$ 以天计。$\tau=4$ 时 $V=4$；$\tau=3$ 时 $V=3.464$。

$$\frac{\partial V}{\partial\tau}=\frac{1}{\sqrt{\tau}}=0.5,$$

所以每过一天，价值约减少 0.5（一阶近似）。这就是时间损耗。

### 3.4 代码：期权的 Theta



```python

import numpy as np
from scipy.special import ndtr

def call(F, K, T, vol, r=0.02):
    if T == 0:
        return max(F - K, 0)
    w = vol * np.sqrt(T)
    d1 = (np.log(F / K) + 0.5 * w * w) / w
    return np.exp(-r * T) * (F * ndtr(d1) - K * ndtr(d1 - w))

F, K, T, vol = 100, 100, 0.5, 0.25
base = call(F, K, T, vol)
theta = (call(F, K, T - 1 / 365, vol) - base) / (1 / 365)
print("期权价值", round(base, 4), "Theta（每年）", round(theta, 4))
print("每过一天约变化", round(theta / 365, 4))
```


**读数与应用**：看涨的 Theta 通常是负的：时间流逝，期权价值下降。临近到期时，Theta 的绝对值会加速变大（时间价值衰减加快）。

**常见错误**：

1. 把 Theta 的符号写反。买方长期权，时间流逝通常亏；卖方赚。
2. 用交易日代替日历日。期权到期按日历日，但波动率按交易日；要统一口径。
3. 把 Theta 当成“每天固定损耗”。它是局部斜率，时间越长、越接近到期，变化越大。

**停下来想一想**：为什么临近到期，Theta 的绝对值会变大？

<details><summary>对照你的思路</summary>

剩余期限越短，期权的时间价值越少，而且衰减速度越快。平值期权在最后几天几乎每天都在快速失去时间价值；深度实值/虚值的 Theta 较小。用 $V=2\sqrt{\tau}$ 举例：$\tau$ 越小，$dV/d\tau=1/\sqrt{\tau}$ 越大。

</details>

### 3.5 全微分的一个例子

$V=100$，$\Delta=0.5$，$\Gamma=0.02$，$\Theta=-0.01$（每天），Vega=20（每个波动率点）。若标的白涨 1、波动率涨 0.5 个点、过了一天：

$$dV\approx0.5\times1+\frac12\times0.02\times1^2+(-0.01)\times1+20\times0.5=0.5+0.01-0.01+10=10.5.$$

这个分解告诉你“哪一项在赚钱/亏钱”。实盘归因就是把这个式子逐项算出来。

## 4. 泰勒展开：把期权损益拆成几笔

### 4.1 一元泰勒

$$f(x+h)\approx f(x)+f'(x)h+\frac12f''(x)h^2+\frac16f^{(3)}(x)h^3+\dots$$

一阶项是“斜率”，二阶项是“曲率”，更高阶是“曲率的变化”。

### 4.2 期权版

$$dV\approx\underbrace{\Delta\,dF}_{\text{方向}}+\underbrace{\frac12\Gamma(dF)^2}_{\text{曲率}}+\underbrace{\Theta\,dt}_{\text{时间}}+\underbrace{\text{Vega}\,d\sigma}_{\text{波动率}}.$$

这就是期权交易员每天看的“希腊字母账”。

### 4.3 手算

$F=100$，$\Delta=0.5$，$\Gamma=0.02$，标的涨 1：

- 一阶：$0.5\times1=0.5$；
- 二阶：$0.5\times0.02\times1^2=0.01$；
- 合计约 0.51。

如果标的涨 5，一阶 2.5，二阶 $0.5\times0.02\times25=0.25$，高阶项开始明显；这就是“大涨时 Delta 近似不够”的原因。

### 4.4 代码：泰勒近似



```python

import numpy as np

F, K, T, vol = 100, 100, 0.5, 0.25
base = call(F, K, T, vol)
h = 0.01
delta = (call(F + h, K, T, vol) - call(F - h, K, T, vol)) / (2 * h)
gamma = (call(F + h, K, T, vol) - 2 * base + call(F - h, K, T, vol)) / h ** 2

for dF in [1.0, 5.0]:
    exact = call(F + dF, K, T, vol) - base
    approx = delta * dF + 0.5 * gamma * dF ** 2
    print("dF", dF, "精确", round(exact, 4), "二阶近似", round(approx, 4), "误差", round(exact - approx, 4))
```


**读数与应用**：$dF=1$ 时，二阶近似已经很准；$dF=5$ 时，误差变大，因为三阶及以上项不可忽略。实盘里，标的大幅跳空时，Delta-Gamma 近似会失效，需要重新定价。

**常见错误**：

1. 用一阶近似处理大涨大跌。Gamma 会让误差迅速扩大。
2. 忘记 Theta 和 Vega。只对冲 Delta，不等于损益归因完整。
3. 在到期日附近用固定 Gamma。临近到期，Gamma 变化很快。

**停下来想一想**：为什么卖方喜欢“标的横盘”，买方喜欢“标的大动”？

<details><summary>对照你的思路</summary>

卖方是负 Gamma：标的不动，Theta 为正（收时间价值），每天赚；标的大动，负 Gamma 让他亏。买方是正 Gamma：标的不动，Theta 为负（付时间价值）；标的大动，正 Gamma 让他赚。这就是“卖方赚时间、买方赚波动”的直觉。

</details>

## 5. 定积分：期望、VaR、尾部损失

### 5.1 积分是面积

连续随机变量 $X$ 的期望：

$$E[X]=\int_{-\infty}^{\infty}x f(x)\,dx.$$

这里的 $f(x)$ 是密度，$f(x)dx$ 是“落在小区间里的概率”。积分就是把所有小区间的“值 × 概率”加起来。

### 5.2 VaR 与 ES

95% VaR 是损失分布的 95% 分位点：

$$P(L\le \text{VaR})=0.95.$$

ES（Expected Shortfall）是超过 VaR 的平均损失：

$$\text{ES}=E[L\mid L>\text{VaR}].$$

**公式拆解**
- **符号**：$L$ 损失（元），$\alpha$ 置信水平；$\text{VaR}$ 是损失分位点，$\text{ES}$ 是超过 VaR 的条件均值。
- **结构**：VaR 是一个点（门槛），ES 是一段（尾部平均），ES 对尾部更敏感。
- **代入**：$L\sim U(0,10)$ → 95% VaR $=9.5$，ES $=(9.5+10)/2=9.75$。
- **边界**：VaR 不满足次可加性、不是最大损失；ES 要求尾部期望存在。
- **用途**：限额、保证金、压力测试；不能用 VaR 比较不同分布的组合。

VaR 回答“最坏 5% 的门槛在哪”，ES 回答“真的跌进最坏 5%，平均亏多少”。ES 对尾部更敏感。

### 5.3 手算

损失 $L$ 在 $[0,10]$ 上均匀分布，密度 $0.1$。

$$P(L>8)=\int_8^{10}0.1\,dx=0.2.$$

$$E[L\mid L>8]=\frac{1}{0.2}\int_8^{10}x\cdot0.1\,dx=\frac{0.1\times(100-64)/2}{0.2}=9.$$

所以 80% 分位的 VaR 是 8，ES 是 9。

### 5.4 代码：正态 VaR/ES



```python

import numpy as np
from scipy.stats import norm
from scipy.integrate import quad

mu, sigma = 0.0, 0.02
alpha = 0.95
q = mu + norm.ppf(1 - alpha) * sigma
var_loss = -q
es_loss = -quad(lambda x: x * norm.pdf(x, mu, sigma), -np.inf, q)[0] / (1 - alpha)

print("95% VaR（损失）", round(var_loss, 4))
print("95% ES（损失）", round(es_loss, 4))
```


**读数与应用**：正态假设下，95% VaR 约 1.645 个标准差，ES 约 2.06 个标准差。真实收益有厚尾，ES 往往比正态估计更大；不要只用 VaR 做风险限额。

**常见错误**：

1. 把密度当概率。$f(x)$ 可以大于 1，概率是积分。
2. 把 VaR 当“最大损失”。VaR 只是分位点，超过它的损失可能很大。
3. 用正态分布估尾部。真实尾部更厚，ES 会被低估。

**停下来想一想**：为什么 ES 比 VaR 更适合做尾部风险指标？

<details><summary>对照你的思路</summary>

VaR 只告诉你“门槛在哪”，不告诉你“超过门槛后有多惨”；ES 是超过门槛的平均损失，包含了尾部的深度。两个组合可以有相同的 VaR，但 ES 差很多。风险管理更关心“真的出事时亏多少”，所以 ES 更合适。

</details>

## 6. Black–Scholes 偏微分方程

### 6.1 从复制到方程

设期权价值 $V(F,t)$。一小段时间内：

$$dV=V_t\,dt+V_F\,dF+\frac12V_{FF}\,dF^2.$$

期货价格在风险中性下：

$$dF=\sigma F\,dW,\qquad dF^2=\sigma^2F^2dt.$$

代入：

$$dV=\left(V_t+\frac12\sigma^2F^2V_{FF}\right)dt+V_F\,dF.$$

构造“卖一份期权、买 $V_F$ 份期货”的组合，消掉 $dF$ 项：

$$d\Pi=\left(V_t+\frac12\sigma^2F^2V_{FF}\right)dt.$$

组合无风险，只能赚无风险利率：

$$d\Pi=r\Pi\,dt=r(V-V_FF)\,dt.$$

两式相等，整理得

$$V_t+\frac12\sigma^2F^2V_{FF}-rV=0.$$

**公式拆解**
- **符号**：$V(F,t)$ 期权价值，$V_t$、$V_{FF}$ 偏导；$\sigma$ 波动率、$r$ 利率，三项单位均为元/年。
- **结构**：时间衰减 $+$ 曲率项 $-$ 资金成本 $=0$；推导中 $dF$ 被消掉，故不含 $\mu$。
- **代入**：$F=K=100$、$\sigma=20\%$、$r=3\%$、$T=1$ → $-3.620+3.852-0.232\approx0$。
- **边界**：$K,T$ 只进边界条件；股票期权多 $rFV_F$ 一项，抄错就偏。
- **用途**：定价、风险归因、校准；不是方向预测，也不是市场价。

这就是期货期权的 BS 方程。股票期权多一项 $rFV_F$，因为股票有资金成本。

### 6.2 每一项在说什么

| 项 | 含义 |
|---|---|
| $V_t$ | 时间流逝的影响（Theta） |
| $\frac12\sigma^2F^2V_{FF}$ | 曲率/波动带来的二阶项（Gamma） |
| $-rV$ | 无风险收益的机会成本 |

### 6.3 代码：验证 PDE



```python

import numpy as np

Ft, eps = 100.0, 0.5
V0 = call(Ft, K, T, vol)
Vt = (call(Ft, K, T - 1 / 365, vol) - V0) / (1 / 365)
VF = (call(Ft + eps, K, T, vol) - call(Ft - eps, K, T, vol)) / (2 * eps)
VFF = (call(Ft + eps, K, T, vol) - 2 * V0 + call(Ft - eps, K, T, vol)) / eps ** 2
residual = Vt + 0.5 * vol ** 2 * Ft ** 2 * VFF - 0.02 * V0

print("V_t, V_F, V_FF", round(Vt, 4), round(VF, 4), round(VFF, 4))
print("PDE 残差（应接近 0）", round(residual, 4))
```


**读数与应用**：残差应该在 0 附近；如果不接近 0，通常是差分步长、时间方向或符号写错。这个检查比重背公式有用：它让你知道每一项到底在算什么。

**常见错误**：

1. 时间方向写反。$V_t$ 和 $V_\tau$ 差一个负号。
2. 把股票期权的 $rFV_F$ 项漏掉/多加。期货期权没有这一项。
3. 边界条件写错。有限差分需要正确的上下边界。

**停下来想一想**：为什么 BS 方程里没有漂移项 $\mu$？

<details><summary>对照你的思路</summary>

因为 Delta 对冲消掉了标的的随机项，也消掉了漂移项；组合的收益只来自无风险利率。这就是“风险中性定价”：真实漂移 $\mu$ 不影响无套利价格。漂移只影响真实世界的期望收益，不影响复制成本。

</details>

### 6.4 边界与终端条件

PDE 需要一个终端条件（到期支付）和两个边界条件（$F=0$、$F=F_{\max}$）：

- 终端：$V(F,T)=\max(F-K,0)$（看涨）；
- 下边界：$V(0,t)=0$；
- 上边界：$V(F_{\max},t)\approx e^{-r(T-t)}(F_{\max}-K)$。

边界不是随便写的：它要满足“深度实值/虚值时的渐进行为”。写错边界，网格内部再准也会被边界污染。

---



## 7. 有限差分：把 PDE 变成网格

### 7.1 网格

把标的价格 $F$ 从 0 到 $F_{\max}$ 切成 $N$ 段，时间从 0 到 $T$ 切成 $M$ 段：

$$F_i=i\Delta F,\qquad \tau_n=n\Delta t.$$

用 $V_i^n$ 表示第 $n$ 个时间层、第 $i$ 个价格点的期权价值。

### 7.2 显式格式

对 BS 方程做显式差分：

$$V_i^{n+1}=V_i^n+\Delta t\left[\frac12\sigma^2F_i^2\frac{V_{i+1}^n-2V_i^n+V_{i-1}^n}{\Delta F^2}-rV_i^n\right].$$

从到期日往前推。边界：

- $F=0$：看涨价值 0；
- $F=F_{\max}$：用深度实值的近似 $e^{-r\tau}(F_{\max}-K)$。

### 7.3 稳定性

显式格式有条件稳定：

$$\Delta t\le\frac{\Delta F^2}{\sigma^2F_{\max}^2}.$$

不满足会震荡/发散。实务中常用隐式或 Crank–Nicolson，它们无条件稳定，但每步要解线性方程组。

### 7.4 代码：显式差分



```python

import numpy as np

K, T, vol, r = 100, 0.5, 0.25, 0.02
Fmax, N, M = 200, 200, 5000
F = np.linspace(0, Fmax, N + 1)
dF, dt = Fmax / N, T / M
V = np.maximum(F - K, 0)

for n in range(M):
    tau = (n + 1) * dt
    Vn = V.copy()
    V[1:-1] = Vn[1:-1] + dt * (
        0.5 * vol ** 2 * F[1:-1] ** 2 * (Vn[2:] - 2 * Vn[1:-1] + Vn[:-2]) / dF ** 2
        - r * Vn[1:-1]
    )
    V[0] = 0.0
    V[-1] = np.exp(-r * tau) * (Fmax - K)

print("有限差分", round(V[N // 2], 4))
print("Black-76", round(call(100, K, T, vol), 4))
```


**读数与应用**：两者应该接近；差距来自网格、边界和离散误差。把 $N$ 加倍、$M$ 加倍，误差通常会下降；如果反而震荡，检查稳定性条件。

**常见错误**：

1. 时间方向反了。从到期日往前推，不是从今天往后推。
2. 边界写错。上边界不是“支付”，而是“支付的现值”。
3. 步长太大，显式格式发散。先算一下稳定性条件。

**停下来想一想**：为什么隐式格式更贵但更稳？

<details><summary>对照你的思路</summary>

隐式格式在每个时间步要解一个三对角线性方程组，计算量比显式大；但它无条件稳定，可以用更大的时间步。显式格式每步只做加减乘除，便宜，但步长受稳定性限制。选择取决于精度、速度和实现复杂度。

</details>

### 7.5 隐式与 Crank–Nicolson

显式格式每步只做加减乘除，但受稳定性限制。隐式格式把未知层放在方程左边，每步解一个三对角方程组：

$$(I-\theta A)V^{n+1}=(I+(1-\theta)A)V^n.$$

$\theta=1$ 是全隐式，$\theta=1/2$ 是 Crank–Nicolson。后者二阶精度、无条件稳定，是实务常用；代价是实现更复杂。

## 8. 微分平滑：为什么不能对噪声直接求导

### 8.1 噪声被放大

二阶差分除以 $h^2$：$h$ 越小，分母越小，噪声被放大得越厉害。对报价数据直接求二阶导，得到的往往是“噪声的曲率”，不是期权的 Gamma。

### 8.2 正则化平滑

找一条平滑曲线 $z$，既贴近原始 $y$，又少一些急弯：

$$\min_z\ ||z-y||^2+\lambda||Dz||^2,$$

其中 $D$ 是二阶差分矩阵。解：

$$(I+\lambda D^TD)z=y.$$

$\lambda$ 越大，曲线越平滑，但偏离原始数据越多。

### 8.3 代码：平滑



```python

import numpy as np

rng = np.random.default_rng(701)
x = np.linspace(0, 1, 80)
y = np.sin(2 * np.pi * x) + 0.1 * rng.normal(size=len(x))
D = np.diff(np.eye(len(x)), 2, axis=0)

for lam in [0.0, 1e-3, 1e-1]:
    z = np.linalg.solve(np.eye(len(x)) + lam * D.T @ D, y)
    print("lambda", lam, "二阶差分波动", round(np.std(np.diff(z, 2)), 4), "拟合误差", round(np.std(z - y), 4))
```


**读数与应用**：$\lambda=0$ 时，$z=y$，拟合误差 0，但二阶差分波动最大；$\lambda$ 增大，曲线更平滑，拟合误差上升。选 $\lambda$ 要用交叉验证，不能只看图“顺眼”。

**常见错误**：

1. 先平滑再回测，但平滑参数用了全样本。要滚动/训练段选参数。
2. 过度平滑，把真实的拐点也抹掉。
3. 对价格平滑后求导，却忘了价格本身是非平稳的；通常对收益/隐波做平滑。

**停下来想一想**：为什么不直接对原始报价求 Gamma？

<details><summary>对照你的思路</summary>

报价有买卖价差、最小跳动、时间不同步；直接二阶差分会把噪声放大成“假 Gamma”。要先清洗/对齐/平滑，再求导。但平滑会引入偏差，所以要保留原始数据、报告平滑参数，并做稳健性检查。

</details>

## 9. 久期与凸性：同样的泰勒，用在债券上

### 9.1 债券价格与收益率

固定利率债券的价格是未来现金流的现值：

$$P(y)=\sum_{t=1}^T\frac{C}{(1+y)^t}+\frac{F}{(1+y)^T}.$$

收益率 $y$ 上升，价格下降。

### 9.2 久期与凸性

久期（修正久期）是价格对收益率的一阶敏感度：

$$D=-\frac{1}{P}\frac{dP}{dy}.$$

凸性是二阶敏感度：

$$C=\frac{1}{P}\frac{d^2P}{dy^2}.$$

价格变化近似：

$$\frac{\Delta P}{P}\approx-D\Delta y+\frac12C(\Delta y)^2.$$

这和期权的 Delta-Gamma 展开是同一个数学结构：一阶是方向，二阶是曲率。

### 9.3 手算

5 年期、票息 5%、面值 100、收益率 5% 的债券，价格约 100。久期约 4.33，凸性约 23.9。收益率上升 1%：

$$\frac{\Delta P}{P}\approx-4.33\times0.01+\frac12\times23.9\times0.0001=-0.0433+0.0012=-0.0421.$$

价格约跌 4.21%，而不是 4.33%——凸性让跌幅小一点。

### 9.4 代码：久期与凸性



```python

import numpy as np

def bond_price(y, T=5, c=5, face=100):
    ts = np.arange(1, T + 1)
    return np.sum(c / (1 + y) ** ts) + face / (1 + y) ** T

y = 0.05
P = bond_price(y)
dy = 1e-4
duration = -(bond_price(y + dy) - bond_price(y - dy)) / (2 * dy) / P
convexity = (bond_price(y + dy) - 2 * P + bond_price(y - dy)) / dy ** 2 / P

print("价格、久期、凸性", round(P, 3), round(duration, 3), round(convexity, 3))
for shift in [0.01, -0.01]:
    exact_pct = (bond_price(y + shift) - P) / P
    approx_pct = -duration * shift + 0.5 * convexity * shift ** 2
    print("收益率变动", shift, "精确%", round(exact_pct * 100, 2), "近似%", round(approx_pct * 100, 2))
```


**读数与应用**：久期近似在收益率小幅变动时够用；变动越大，凸性项越重要。债券的“久期/凸性”和期权的“Delta/Gamma”是同一套泰勒展开，只是输入变量不同。

**常见错误**：

1. 把麦考利久期和修正久期混用。修正久期 = 麦考利久期 / $(1+y)$。
2. 只用久期，忽略凸性。利率大幅变动时，凸性项不能省。
3. 假设收益率曲线平行移动。真实曲线可能变陡/变平，单因子久期不够。

**停下来想一想**：为什么债券的凸性通常是正的，而期权的 Gamma 可正可负？

<details><summary>对照你的思路</summary>

普通债券的价格-收益率关系是凸的（下凸），所以凸性为正；利率上升和下降对价格的二阶影响都是“涨多跌少”。期权的 Gamma 取决于头寸：买方正 Gamma，卖方负 Gamma。数学结构一样，符号由现金流/头寸决定。

</details>

### 9.5 久期的三种口径

- 麦考利久期：现金流到期时间的加权平均，单位是年；
- 修正久期：$D_{\text{mod}}=D_{\text{Mac}}/(1+y)$，直接用于价格敏感度；
- 美元久期：$D_{\text{mod}}\times P$，单位是“价格/收益率”。

报告时要写清是哪一种。把麦考利久期直接乘收益率变化，会高估/低估敏感度。

| 泰勒 | $f(x+h)\approx f+f'h+\frac12f''h^2$ | 损益近似 |
| 期望 | $E[X]=\int xf(x)dx$ | 连续分布 |
| ES | $E[L\mid L>\text{VaR}]$ | 尾部损失 |
| BS PDE | $V_t+\frac12\sigma^2F^2V_{FF}-rV=0$ | 期权定价 |
| 久期 | $D=-\frac1P\frac{dP}{dy}$ | 利率一阶风险 |
| 凸性 | $C=\frac1P\frac{d^2P}{dy^2}$ | 利率二阶风险 |

### 八句话

1. 导数是斜率，不是变化量；变化量还要乘输入变动。
2. Gamma 是 Delta 的变化率；平值附近最大。
3. Theta 是时间斜率；买方通常为负，卖方为正。
4. 泰勒展开把损益拆成方向、曲率、时间、波动率四笔。
5. 大涨大跌时，Delta-Gamma 近似会失效。
6. BS 方程是“复制组合无风险”的数学表达。
7. 有限差分把 PDE 变成网格；显式格式有稳定性条件。
8. 久期/凸性和 Delta/Gamma 是同一套泰勒，只是变量不同。

## 本章习题

**1.** $f(x)=x^3$，在 $x=2$ 处用中心差分求导（$h=10^{-4}$），和解析导数比较。

**2.** 期权在 $F=99,100,101$ 时价值 $4.55,5.00,5.55$。求 Delta、Gamma。

**3.** $V=2\sqrt{\tau}$，$\tau=9$ 天。求 $\partial V/\partial\tau$，并估计每过一天的价值变化。

**4.** $\Delta=0.6$，$\Gamma=0.03$，标的涨 2。用二阶泰勒估计期权变化；如果实际变化是 1.32，误差多少？

**5.** 损失在 $[0,10]$ 上均匀。求 $P(L>8)$、$E[L\mid L>8]$。

**6.** BS 方程 $V_t+\frac12\sigma^2F^2V_{FF}-rV=0$ 中，哪一项对应 Theta、哪一项对应 Gamma？

**7.** 显式有限差分中，如果 $\Delta t$ 太大，会发生什么？为什么？

**8.** 5 年期、票息 5%、面值 100、收益率 5% 的债券，久期约 4.3、凸性约 22。收益率上升 1%，用久期+凸性估计价格变化百分比。

<details><summary>参考答案</summary>

**1.** 解析 $3x^2=12$；中心差分 $(2.0001^3-1.9999^3)/(0.0002)\approx12.000$。

**2.** $\Delta=(5.55-4.55)/2=0.50$；$\Gamma=(5.55-2\times5.00+4.55)/1=0.10$。

**3.** $\partial V/\partial\tau=1/\sqrt{\tau}=1/3\approx0.333$；每过一天约减少 0.333。

**4.** 二阶近似 $0.6\times2+0.5\times0.03\times4=1.2+0.06=1.26$；误差 $1.32-1.26=0.06$。

**5.** $P(L>8)=0.2$；$E[L\mid L>8]=9$。

**6.** $V_t$ 对应 Theta；$\frac12\sigma^2F^2V_{FF}$ 对应 Gamma 项。

**7.** 会震荡/发散。显式格式有条件稳定，$\Delta t$ 超过 $\Delta F^2/(\sigma^2F_{\max}^2)$ 时，误差被逐步放大。

**8.** $-4.3\times0.01+0.5\times22\times0.0001=-0.043+0.0011=-0.0419$，约跌 4.19%。
