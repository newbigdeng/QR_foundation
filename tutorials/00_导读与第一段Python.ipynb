## 3. 每一章怎么读

教材章按“十块”组织：

1. **场景**：一个具体的人、一笔具体的交易、一个具体的问题；
2. **前置**：学这一节之前必须会的 2–3 件事；
3. **定义**：符号表 + 一句话定义 + 一个反例；
4. **图解**：时间线、现金流、矩阵形状、反馈回路；
5. **推导**：每一步为什么，至少 3–5 行，不许跳步；
6. **手算例题**：数字代入，算到最后一个数；
7. **代码**：15–30 行，逐段注释，和公式一一对应；
8. **结果解读**：在交易里对应什么，什么情况下失效；
9. **常见错误**：3 条，每条给错误做法和正确做法；
10. **练习 + 一句话小结**：2–3 道题，答案折叠。

举个具体的例子。02 的“期望”一节，先让你算十张票的平均盈亏；再把这个平均写成 $E[X]=\sum p_ix_i$；然后问“胜率 60% 为什么还会亏”；最后用三行代码验证。你读完这一节，应该能自己算一笔新的交易账，而不是记住“期望”两个字的定义。

建议每章留三样东西：一张手算纸、一段可运行代码、一句“这个结论什么时候失效”。这三样比“读完”更能说明你学会了。

**读法**：

- 第一遍：盖住代码，用纸笔做手算例题；做不出来再看“纸笔推演”；
- 第二遍：运行代码，把每个变量和公式里的符号对上；
- 第三遍：做章末习题，错题回到对应小节；
- 第四遍：改一个假设（成本、参数、样本段），看结论什么时候失效。

**不要**：只运行代码不看文字；只看答案不做题；把“看懂了”当成“会算了”。

## 8. 常见问题

**数学基础差，能学吗？**
能，但要从 02 开始，并且用纸笔做例题。不要跳过手算直接看代码；代码是验证，不是替代。

**不会 Python，能学吗？**
能。02 的代码从数组和循环开始；先照着改数字，再自己写。遇到不懂的函数，用 `help()` 或查文档。

**没有数据，能学吗？**
10、16、18 带真实 RB 分钟数据；其他章节用合成数据，目的是讲清机制。学完后，把自己的数据换进去，才是真正的练习。

**要多久？**
零基础全程约 6 个月；因子速成 6–8 周；期权专线 8 周；WT 工程 4–6 周。时间取决于每天投入和是否动手做题。

**学完相当于 QR 什么水平？**
大约相当于“能独立完成初级因子研究、能读懂期权报价、能在 WT-plus 里跑通验证”的入门水平。距离独立管理资金、设计复杂策略，还需要实盘经验、风险管理和更多数学。

**要不要先学 C++ 或机器学习？**
先把 Python 和统计打牢。C++ 是 WT 策略、低延迟执行需要的能力；机器学习是第11章之后的内容。不要一开始就同时学三样，容易每样都停在“hello world”。

**教程里的代码能直接实盘吗？**
不能。教程用合成数据或历史片段讲机制，没有接实时行情、风控、柜台。实盘前要补：数据校验、断线重连、保证金/强平、异常订单处理。
