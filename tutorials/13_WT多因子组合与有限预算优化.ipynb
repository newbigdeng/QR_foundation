## 1. 合成分数：先统一方向，再谈权重

### 1.1 方向

- 动量：越高越看多，方向 +1；
- 拥挤度：越高越危险，方向 −1；
- 库存：越高越看空，方向 −1。

不统一方向，直接相加会把“看多”和“看空”抵消。

### 1.2 标准化/排名

- z-score：保留幅度，对异常值敏感；
- 排名：稳健，但丢幅度；
- 截面标准化：每天做，不用全样本。

### 1.3 权重

- 等权：简单稳健；
- IC 加权：按历史 IC；
- 风险平价：按风险贡献；
- 机器学习：非线性，但容易过拟合。

### 1.4 代码：方向统一 + 排名



```python

import numpy as np
import pandas as pd

momentum = np.array([3.0, 2.0, 1.0])
crowded = np.array([3.0, 1.0, 2.0])

score = 0.5 * (pd.Series(momentum).rank() + pd.Series(4 - crowded).rank())
print("统一方向后的合成分数", score.to_list())
```


**读数与应用**：动量排名 (3,2,1)，拥挤度取反后排名 (1,3,2)，合成 (4,5,3)/2 = (2,2.5,1.5)。方向统一后，分数越高越看多。

**常见错误**：

1. 不统一方向。
2. 用全样本标准化。
3. 权重拍脑袋，不做敏感性。

**停下来想一想**：为什么“排名”比“z-score”更适合多因子合成？

<details><summary>对照你的思路</summary>

不同因子的分布/异常值不同；排名把每个因子压到同一尺度，稳健、可比。z-score 保留幅度，但一个极端值会主导。实务中两者都可用，要报告选择。

</details>

## 2. 组合对照：新因子有没有增量

### 2.1 同日净额

新旧组合在同一交易日比较净收益差，再按日期统计；不要用不同日期/不同市场状态硬比。

### 2.2 基准

至少加两个基准：

- 旧组合（不含新因子）；
- 等波动/等仓位基准（排除“只是加杠杆”）。

### 2.3 代码：同日增量



```python

old = np.array([8.0, -3.0, 5.0]) - 1.0
new = np.array([11.0, 1.0, 8.0]) - 3.0

print("每日净增益", new - old, "合计", round((new - old).sum(), 2))
print("旧均值/新均值", round(old.mean(), 2), round(new.mean(), 2))
```


**读数与应用**：新组合每日增益 (1,2,1)，合计 4；但新组合成本更高（−3 vs −1），要确认增益不是“少交易/加杠杆”造成的。

**常见错误**：

1. 不同日期比较。
2. 只看收益，不看风险/换手。
3. 不做“旧组合/等波动”基准。

**停下来想一想**：新组合收益更高，是否一定说明新因子有用？

<details><summary>对照你的思路</summary>

不一定。可能只是加了杠杆、换了仓位、少交易，或者样本期特殊。要同日配对、等波动基准、扣费、样本外，才能说“增量”。

</details>

## 3. 边际风险：谁在主导组合

### 3.1 贡献

组合方差 $w^TSw$，边际贡献：

$$\frac{\partial\sigma^2}{\partial w_i}=2(Sw)_i.$$

**公式拆解**
- **符号**：$w_i$ 权重（可正负），$S$ 协方差矩阵（收益率²），$\sigma^2=w^TSw$ 组合方差。
- **结构**：$\partial\sigma^2/\partial w_i=2(Sw)_i$，$(Sw)_i$ 是资产与组合的协方差，不是自身方差。
- **代入**：$w=(0.8,0.2)$、$S=\mathrm{diag}(1,16)$ → $Sw=(0.8,3.2)$、梯度 $(1.6,6.4)$。
- **边界**：$(Sw)_i=0$ 时无边际风险；对波动率求导是 $(Sw)_i/\sigma$。
- **用途**：风险预算、加仓判断；不是收益贡献，也不是手数。

风险贡献占比：

$$RC_i=\frac{w_i(Sw)_i}{w^TSw}.$$

**公式拆解**
- **符号**：$RC_i$ 风险贡献占比（无量纲，$\sum RC_i=1$），$w_i(Sw)_i$ 为该资产对组合方差的贡献。
- **结构**：$RC_i=w_i(Sw)_i/w^TSw$，权重×协方差再除以组合方差，Euler 分解保证加总为 1。
- **代入**：$w\cdot Sw=(0.64,0.64)$、$\sigma^2=1.28$ → $RC=(50\%,50\%)$，尽管权重 80/20。
- **边界**：$\sigma^2=0$ 无定义；负权重时 $RC_i$ 可为负、单项可超 100%。
- **用途**：风险预算/风险归因；不是收益贡献，也不等于“要减仓”。

### 3.2 代码：风险贡献



```python

import numpy as np

w = np.array([0.8, 0.2])
S = np.diag([1.0, 16.0])
contribution = w * (S @ w)
print("方差贡献", contribution, "占比", (contribution / contribution.sum()).round(3))
```


**读数与应用**：20% 的仓位贡献 50% 的风险；权重和风险贡献不是一回事。

**常见错误**：

1. 按权重分配风险预算。
2. 忽略相关性。
3. 不做压力测试。

**停下来想一想**：为什么低波动资产也可能贡献很多风险？

<details><summary>对照你的思路</summary>

若它和组合其他部分高度相关，或权重很大，边际贡献 $w_i(Sw)_i$ 仍可能高。风险贡献看“协方差 × 权重”，不只看自身波动。

</details>

## 4. 有限预算：从分数到仓位

### 4.1 约束

- 预算：$\sum|w_i|\le1$；
- 单品种上限；
- 换手上限；
- 保证金/整数手数。

### 4.2 求解

用 SLSQP/凸优化；先连续，再取整，再检查约束。

### 4.3 代码：有限预算



```python

from scipy.optimize import minimize

mu = np.array([0.04, 0.08, 0.12])
cov = np.diag([0.05 ** 2, 0.12 ** 2, 0.25 ** 2])
res = minimize(lambda x: x @ cov @ x, [1 / 3] * 3, bounds=[(0, 0.5)] * 3,
               constraints=[{"type": "eq", "fun": lambda x: x.sum() - 1}], method="SLSQP")
print("权重", res.x.round(3), "波动", round(np.sqrt(res.fun), 4))
```


**读数与应用**：单品种上限 0.5 时，权重被约束改变；取整后还要检查预算/风险。

**常见错误**：

1. 连续权重直接当手数。
2. 约束只在回测最后加。
3. 取整后不复查。

**停下来想一想**：为什么“有限预算”会改变最优因子权重？

<details><summary>对照你的思路</summary>

预算/上限让可行集变小，优化器不能把所有资金压在高收益因子；它会在约束边界上重新分配。约束越紧，最优越偏离无约束解。

</details>

## 本章小结

### 一张表

| 步骤 | 做法 | 检查 |
|---|---|---|
| 方向 | ±1 | 不抵消 |
| 标准化 | 排名/z-score | 截面、不用全样本 |
| 合成 | 等权/IC/风险 | 敏感性 |
| 对照 | 同日、旧组合、等波动 | 增量 |
| 风险 | $RC_i$ | 不是权重 |
| 执行 | 约束/整数手数 | 复查 |

### 八句话

1. 先统一方向，再谈权重。
2. 排名稳健，z-score 保留幅度。
3. 新因子要有同日/等波动基准。
4. 收益增量要扣费/看换手。
5. 风险贡献不等于权重。
6. 低波动资产也可能贡献很多风险。
7. 约束会改变最优权重。
8. 连续权重必须落到整数手数并复查。

<details><summary>参考答案</summary>

**1.** 动量 +1；拥挤度 −1。

**2.** 拥挤度取反排名 (1,3,2)，合成 (4,5,3)/2=(2,2.5,1.5)。

**6.** 可行集变小，权重被推到约束边界，偏离无约束解。

**7.** 排除“只是加杠杆/加仓位”带来的收益。

**8.** 预算、风险、保证金、换手。

</details>
