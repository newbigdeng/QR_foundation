
## 在量化研究中的位置

- **谁用**：组合经理、研究员、业绩归因。
- **输入**：新/旧组合收益、基准、成本。
- **输出**：beta/alpha、残差、分状态增量、归因表。
- **在流程的哪一步**：业绩归因与增量评估。
- **失败模式**：把 beta 当本事、不同日比较、不扣成本。
- **最小工作流**：基准回归 → 残差 → 同日配对 → 分状态 → 扣费 → 归因。

## 1. 对基准回归：扣掉共同成分

### 1.1 模型

$$r_{\text{new}}=\alpha+\beta r_{\text{old}}+\epsilon.$$

- $\beta$：新组合对旧组合的暴露；
- $\epsilon$：扣掉旧组合后的增量；
- $\alpha$：平均增量（若显著）。

**公式拆解**
- **符号**：$r_{new},r_{old}$ 新旧收益（%），$\alpha$ 截距、$\beta$ 暴露、$\epsilon$ 残差。
- **结构**：OLS 拆成 $\beta r_{old}$（暴露）与 $\alpha+\epsilon$（增量）。
- **代入**：$r_{old}=3\%$、$\beta=1.2$、$r_{new}=5\%$ → 暴露 3.6%、$\alpha=1.4\%$、扣费后 1.0%。
- **边界**：$\beta>1$ 可能是杠杆；$r_{old}$ 方差 0 则 $\beta$ 不可估。
- **用途**：业绩归因起点；$\alpha$ 不是因子贡献或风险。

### 1.2 代码



```python

import numpy as np

base = np.array([-2.0, 0.0, 2.0])
new = np.array([-2.0, 1.0, 4.0])

beta = ((base - base.mean()) @ (new - new.mean())) / np.sum((base - base.mean()) ** 2)
residual = new - beta * base
print("beta", round(beta, 3), "残差", residual.round(3))
```


**读数**：beta 1.5，残差 (−0.5, 1, 0.5)；新组合有一部分只是“放大旧组合”。

**常见错误**：

1. 不扣基准，直接比收益。
2. 用价格代替收益。
3. 忽略 beta 的时变性。

---

## 2. 同日配对差

### 2.1 做法

同一天，新组合 − 旧组合；再对日差做均值/区间。

### 2.2 代码



```python

old = np.array([2.0, -5.0, 3.0])
new2 = np.array([1.0, -2.0, 2.0])
diff = new2 - old
print("配对差", diff, "均值", round(diff.mean(), 3))
```


**读数**：配对差 (−1,3,−1)，均值 0.33；方向不稳定。

**常见错误**：

1. 不同日期比较。
2. 只看均值，不看逐日。
3. 不扣成本。

---

## 3. 分状态看增量

### 3.1 为什么要分

新因子可能只在上涨/下跌/震荡日有用；平均可能掩盖。

### 3.2 代码



```python

base_days = np.array([1.0, -2.0, 3.0, -1.0, 2.0, -3.0])
diff_days = np.array([0.5, 1.0, -0.5, 0.8, -0.2, 1.2])

up = base_days >= 0
print("上涨日平均增量", round(diff_days[up].mean(), 3))
print("下跌日平均增量", round(diff_days[~up].mean(), 3))
```


**读数**：上涨日 0.1、下跌日 1.0；增量主要来自下跌日（保护）。

**常见错误**：

1. 只看全样本平均。
2. 分组样本太少还下结论。
3. 不区分“保护”与“进攻”。

---

## 4. 案例：把 5% 的增量拆开

新组合年化多赚 5%，先别归功于因子。拆成：

| 来源 | 算法 | 例子 |
|---|---|---|
| 市场暴露 | $\beta\times r_{\text{old}}$ | 1.2×3%=3.6% |
| 选择/因子 | $\alpha$ | 5%−3.6%=1.4% |
| 成本 | 换手×费率 | −0.4% |
| 净增量 | $\alpha-$成本 | 1.0% |

### 代码：增量拆解



```python

total_new, beta, base_return, cost = 5.0, 1.2, 3.0, 0.4
alpha = total_new - beta * base_return
print("市场暴露贡献", round(beta * base_return, 2))
print("alpha", round(alpha, 2), "扣费后净增量", round(alpha - cost, 2))
```


**读数**：5% 里 3.6% 只是市场暴露，alpha 1.4%，扣费后 1.0%。**不拆开，就会把 beta 当本事。**

## 5. 风险调整后的增量

只看收益增量不够，还要看“单位风险”的增量：

$$\Delta\text{Sharpe}=\frac{\mu_{\text{new}}-\mu_{\text{old}}}{\sigma_{\text{diff}}}.$$

**公式拆解**
- **符号**：$\mu_{new},\mu_{old}$ 平均收益，$\sigma_{diff}$ 配对差标准差。
- **结构**：$\Delta\text{Sharpe}=\frac{\mu_{new}-\mu_{old}}{\sigma_{diff}}$：平均增量÷增量波动，同日相减消共同波动。
- **代入**：$7\%-5\%=2\%$、$\sigma_{diff}=8\%$ → 0.25。
- **边界**：$\sigma_{diff}=0$ 退化无定义；配对差自相关时 $t$ 被高估。
- **用途**：风险调整后的增量；不是年化夏普，也不是胜率。

配对差的标准误：

$$SE=\frac{s_{\text{diff}}}{\sqrt T}.$$

**公式拆解**
- **符号**：$s_{diff}$ 配对差样本标准差，$T$ 独立配对期数，$SE$ 均值标准误。
- **结构**：$SE=s_{diff}/\sqrt T$，样本翻 4 倍误差减半；$t=\bar d/SE=\Delta\text{Sharpe}\sqrt T$。
- **代入**：$s_{diff}=8\%$、$T=64$ → $SE=1\%$；$\bar d=2\%$ → $t=2.0$。
- **边界**：$T=1$ 无定义，$s_{diff}=0$ 时 $t$ 发散；重叠样本令 $t$ 虚高。
- **用途**：给增量设区间、估样本量；$t$ 大不等于可交易。

### 代码：配对差的 t 值



```python

import numpy as np

diff = np.array([1.0, 2.0, 1.0, -0.5, 1.5, 0.8])
t_stat = diff.mean() / (diff.std(ddof=1) / np.sqrt(len(diff)))
print("配对差均值", round(diff.mean(), 3), "t 值", round(t_stat, 3))
```


**读数**：配对差均值 0.97、t 值约 2.6；方向为正，但样本只有 6 天，不能下强结论。

**常见错误**：

1. 只看收益增量，不看波动。
2. 用独立样本 t 检验，忽略同日相关。
3. 样本太短就宣称“显著”。

## 本章小结

### 一张表

| 方法 | 做法 | 用途 |
|---|---|---|
| 回归残差 | $r_{new}=\alpha+\beta r_{old}+\epsilon$ | 扣共同成分 |
| 配对差 | 同日 new−old | 增量 |
| 分状态 | 上涨/下跌日 | 保护/进攻 |
| 成本 | 扣费后 | 净增量 |

### 六句话

1. 不扣基准，增量可能是杠杆。
2. beta 说明新组合对旧组合的暴露。
3. 残差才是增量。
4. 同日配对，避免日期效应。
5. 分状态看，平均会掩盖。
6. 增量要扣成本。
