## 1. 两状态支付表与复制

### 1.1 支付矩阵

两个可能世界：上涨、下跌。两种资产：债券、股票。

| 资产 | 今天价格 | 上涨支付 | 下跌支付 |
|---|---:|---:|---:|
| 债券 | 1 | 1 | 1 |
| 股票 | 100 | 120 | 90 |

矩阵：

$$A=\begin{pmatrix}1&120\\1&90\end{pmatrix},\qquad p=\begin{pmatrix}1\\100\end{pmatrix}.$$

每一列是一种资产，每一行是一个世界。

### 1.2 复制

要用 $h_b$ 份债券、$h_s$ 份股票复制一个支付 $x=(x_u,x_d)$：

$$A h=x.$$

若 $A$ 可逆，$h=A^{-1}x$，复制成本 $p^Th$。

**公式拆解**
- **符号**：$A$ 支付矩阵（行=状态、列=资产），$h$ 复制份数，$x$ 目标支付，$p$ 今天价格。
- **结构**：$Ah=x$，解 $h=A^{-1}x$，复制成本 $p^Th$。
- **代入**：$A=\begin{pmatrix}1&120\\1&90\end{pmatrix}$，$x=(20,0)$：$h=(-60,2/3)$，成本 $6.67$，两状态恰好复制。
- **边界**：$\det A=0$ 时不能唯一复制，只能给价格区间。
- **用途**：复制定价与离散对冲；行列别搞反，真实概率不进 $A$。

### 1.3 代码：复制看涨



```python

import numpy as np

A = np.array([[1.0, 120.0], [1.0, 90.0]])
price = np.array([1.0, 100.0])

payoff = np.maximum(A[:, 1] - 100.0, 0)   # K=100 看涨
h = np.linalg.solve(A, payoff)
print("债券/股票份数", h.round(3))
print("复制成本", round(price @ h, 3))
```


**读数与应用**：看涨支付 (20,0)，复制需要债券/股票份数；复制成本就是无套利价格。

**常见错误**：

1. 把行/列搞反。
2. 忘记“今天价格”和“支付”是两套数。
3. 用真实概率算复制成本。

**停下来想一想**：为什么复制成本就是“无套利价格”？

<details><summary>对照你的思路</summary>

若市场价格高于复制成本，卖期权、买复制组合可锁定无风险利润；若低于，买期权、卖复制组合。套利会把价格推回复制成本。所以复制成本是唯一无套利价格（在可复制、无摩擦假设下）。

</details>

## 2. 状态价格

### 2.1 定义

设 $q_u,q_d$ 是“上涨/下跌状态”的状态价格：今天买 1 元“只在该状态支付”的证券，要付 $q$。

任何支付 $x$ 的现值：

$$\text{PV}=q_ux_u+q_dx_d.$$

债券/股票价格也必须满足：

$$q_u+q_d=1,\qquad 120q_u+90q_d=100.$$

### 2.2 解

$$q_u=\frac13,\qquad q_d=\frac23.$$

这是风险中性概率（无息时）；状态价格 = 风险中性概率 × 贴现因子。

**公式拆解**
- **符号**：$q_u,q_d$ 为状态价格，$x_u,x_d$ 为状态支付，PV 为现值。
- **结构**：$\text{PV}=q_ux_u+q_dx_d$；债券与股票给出 $q_u+q_d=1$、$120q_u+90q_d=100$。
- **代入**：解得 $q_u=1/3,q_d=2/3$；看涨支付 $(20,0)$ 的价格 $=20/3=6.67$。
- **边界**：$q\ge0$ 才无套利；无息时和为 1，有息时和为 $e^{-rT}$。
- **用途**：给所有支付定价、构造风险中性概率；$q$ 不是真实概率。

### 2.3 代码：解状态价格



```python

q = np.linalg.solve(A.T, price)
print("状态价格", q.round(4))
print("看涨价格", round(q @ payoff, 4), "复制成本", round(price @ h, 4))
```


**读数与应用**：$q=(1/3,2/3)$；看涨价格 $20/3\approx6.67$，和复制成本一致。状态价格为正，是无套利的条件。

**常见错误**：

1. 状态价格之和不为 1（无息时）。
2. 出现负状态价格，仍有套利。
3. 把状态价格当真实概率。

**停下来想一想**：状态价格为什么必须为正？

<details><summary>对照你的思路</summary>

若某个状态价格为负，意味着“持有一个在该状态支付为正的证券，今天还能收钱”；这是套利。无套利要求所有状态价格为正。

</details>

## 3. 不完备市场与价格区间

### 3.1 不完备

若可能世界数 > 独立可交易资产数，支付矩阵不是方阵/不满秩，不能复制所有支付。

### 3.2 价格区间

所有“正的状态价格、且与可交易资产价格一致”的集合，给出新资产的上下界：

$$C_{\min}\le C\le C_{\max}.$$

不同状态价格权重都能给已有资产正确定价，但给新期权不同价格。

### 3.3 代码：三状态、两资产



```python

# 三状态、两资产：股票 80/100/120，债券 1/1/1，无息，股票今天 100
states = np.array([80.0, 100.0, 120.0])
q = np.array([0.25, 0.5, 0.25])   # 一组与股票/债券一致的状态价格
print("状态价格和", q.sum(), "股票定价", q @ states)
print("看涨 K=100 价格", q @ np.maximum(states - 100, 0))
```


**读数与应用**：这组 $q$ 给股票定价 100，但换成另一组 $q$（同样满足约束）会给看涨不同价格；不完备市场只能给区间。

**常见错误**：

1. 用一组 $q$ 就宣称“唯一价格”。
2. 忘记检查 $q>0$。
3. 把“不可复制”当成“没有价格”。

**停下来想一想**：不完备市场里，期权为什么还有价格区间？

<details><summary>对照你的思路</summary>

因为无套利只要求“存在正状态价格与可交易资产一致”；不同的状态价格给新期权不同现值，但都不可被套利。区间是所有可行 $q$ 下的最小/最大现值。

</details>

## 4. 随机折现因子与风险中性

### 4.1 随机折现因子

设真实概率 $p=(p_u,p_d)$，状态价格 $q$。定义：

$$m=\frac{q}{p}.$$

则任何支付 $x$ 的现值：

$$\text{PV}=E[m x].$$

$m$ 叫随机折现因子（SDF）：坏状态 $m$ 大（钱更值钱），好状态 $m$ 小。

**公式拆解**
- **符号**：$p$ 真实概率，$q$ 状态价格，$m=q/p$ 为随机折现因子。
- **结构**：$\text{PV}=E[mx]=\sum_ip_im_ix_i$，把状态价格定价改写成真实概率下的期望。
- **代入**：$p=(0.5,0.5),q=(1/3,2/3)$：$m=(2/3,4/3)$，$E[m]=1$；看涨 $PV=0.5\times(2/3\times20)=6.67$。
- **边界**：$E[m]=1$ 只在无息时成立；有息时 $E[m]$ 等于零息债价格。
- **用途**：资产定价的统一语言、拆风险溢价；$m$ 不是概率。

### 4.2 风险中性

若 $q$ 之和为 1（无息），$q$ 就是风险中性概率；真实概率 $p$ 与 $q$ 的差，就是风险溢价。

### 4.3 代码：SDF



```python

p = np.array([0.5, 0.5])
q2 = np.array([1 / 3, 2 / 3])
m = q2 / p
print("SDF", m.round(3), "E[m]", round(p @ m, 3))
```


**读数与应用**：$m=(2/3,4/3)$，坏状态（下跌）的 $m$ 更大；$E[m]=1$（无息）。这就是“坏日子的一块钱更贵”。

**常见错误**：

1. 把 $q$ 当 $m$。
2. 忘记除以真实概率。
3. 在有利率时忘记贴现。

**停下来想一想**：为什么 $E[m]=1$（无息）？

<details><summary>对照你的思路</summary>

无息债券今天 1，任何状态支付 1；现值 $E[m\times1]=1$。所以 $E[m]=1$。有息时，$E[m]=1/(1+r)$ 或按贴现因子调整。

</details>

## 本章小结

### 一张表

| 概念 | 公式 | 用途 |
|---|---|---|
| 支付矩阵 | $A$ | 复制 |
| 复制 | $Ah=x$ | 定价 |
| 状态价格 | $q=A^{-T}p$ | 现值 |
| 不完备 | 方程少于未知数 | 区间 |
| SDF | $m=q/p$ | 风险溢价 |

### 八句话

1. 每一列一种资产，每一行一个世界。
2. 复制成本 = 无套利价格。
3. 状态价格之和为 1（无息）。
4. 状态价格必须为正。
5. 不完备市场只能给区间。
6. 不同 $q$ 都能给已有资产正确定价。
7. SDF = 状态价格 / 真实概率。
8. 坏状态的钱更值钱。
